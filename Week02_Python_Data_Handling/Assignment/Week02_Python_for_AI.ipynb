# 1. Variables

## What is a Variable?

A variable is a name used to store a value in a Python program.

Instead of repeatedly writing a value, we can store it inside a
variable and use the variable whenever we need that value.

### Syntax

variable_name = value

For example:

name = "John"
age = 25
salary = 50000

Here:

- `name` stores the value `"John"`.
- `age` stores the value `25`.
- `salary` stores the value `50000`.

The `=` symbol is called the assignment operator. It assigns the
value on the right side to the variable on the left side.

Variables are important in AI and data handling because programs
need to store information such as names, ages, prices, scores,
features, predictions, and other data.

```python
name = "John"
age = 25
salary = 50000
skills = ["Python", "AI", "Salesforce"]

employee = {
    "name": "John",
    "age": 25,
    "skill": "Python"
}
```

## 1. Functions

### What is a Function?

A function is a reusable block of Python code designed to perform
a specific task.

Instead of writing the same code repeatedly, we can place the code
inside a function and call the function whenever we need it.

### Basic Structure

def function_name():
    statements

The `def` keyword is used to define a function.

### Why are Functions Important?

Functions help us:

- Reuse code
- Avoid repetition
- Organize programs
- Make code easier to understand

## 2. Parameters

### What is a Parameter?

A parameter is a variable defined inside a function's parentheses.
It allows us to pass information into the function.

For example:

def greet(name):

Here, `name` is a parameter.

When we call the function, we provide an actual value:

greet("Jayasri")

The value `"Jayasri"` is passed to the `name` parameter.

### Why are Parameters Useful?

Parameters make functions flexible.

Instead of creating separate functions for different values, we can
create one function and provide different values whenever we call it.
This is one of the main advantages of functions: the same logic can
work with different input values.

```python
def greet(name):
    print("Hello", name)

greet("Jayasri")
greet("John")
greet("Priya")
```

## 3. Multiple Parameters

A function can accept more than one parameter.

For example, a salary calculation may require:

- Basic salary
- Bonus

We can pass both values to a function.

The order of the arguments should match the order of the parameters
unless keyword arguments are used.

```python
def calculate_salary(salary, bonus):
    return salary + bonus

result = calculate_salary(50000,5000)

print(result)
```

## 4. Return Values

### What is `return`?

The `return` statement sends a value from a function back to the
part of the program where the function was called.

For example:

def add(a, b):
    return a + b

The function calculates the result and returns it.

We can store that returned value in another variable.

### `print()` vs `return`

`print()` displays a value on the screen.

`return` sends a value back so that it can be stored, reused,
or processed further.

This distinction is very important when creating reusable functions.

```python
def add(a, b):
    return a + b

result = add(10, 20)

print("Result:", result)
```

## 5. List Comprehension

### What is List Comprehension?

List comprehension is a concise way of creating a new list from
an existing collection.

Instead of writing several lines using a loop, we can often create
the same list in a single line.

### Normal For Loop

For example, to calculate the squares of numbers:

numbers = [1, 2, 3, 4, 5]

squares = []

for x in numbers:
    squares.append(x * x)

### Using List Comprehension

The same operation can be written as:

squares = [x * x for x in numbers]

List comprehension is commonly used for transforming or filtering
data.

```python
numbers = [1, 2, 3, 4, 5]

squares = [x * x for x in numbers]

print("Numbers:", numbers)
print("Squares:", squares)
```

## 6. List Comprehension with Conditions

List comprehension can also include a condition.

For example, we can create a list containing only even numbers.

The general structure is:

[expression for item in collection if condition]

This allows us to transform and filter data at the same time.

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = [x for x in numbers if x % 2 == 0]

print(even_numbers)
```

## 7. Exception Handling

### What is an Exception?

An exception is an error that occurs while a Python program is
running.

For example, trying to divide a number by zero causes an error.

Without exception handling, the program may stop when the error occurs.

### Why is Exception Handling Important?

Exception handling allows us to handle errors gracefully instead
of allowing the entire program to crash.

Python commonly uses:

- `try`
- `except`
- `else`
- `finally`

The basic structure is:

#### try:
contains the code that might produce an error.

#### except tells Python:
"If an error happens in the try block, do this instead."

#### else: 
runs only when the try block succeeds without any error.

#### finally:
contains code that should run whether an error happens or not.

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("You cannot enter zero.")

else:
    print("Result:", result)

finally:
    print("Program execution completed.")
```

```python
try:
    file = open("data.txt", "r")
    data = file.read()

except FileNotFoundError:
    print("File not found.")

else:
    print("File read successfully.")

finally:
    print("Finished file operation.")
```

## 8. Functions with Lists

Functions can receive lists as parameters.

For example, a function can receive a list of numbers and calculate
the total.

Example:

numbers = [10, 20, 30]

A function can process this list and return:

60

This approach is useful because the same function can work with
different lists of numbers.

```python
def calculate_total(numbers):
    return sum(numbers)

numbers = [10, 20, 30]

total = calculate_total(numbers)

print("Total:", total)
```

### Functions Required

`calculate_total()`
- Calculates the sum of all numbers.

`calculate_average()`
- Calculates the average of the numbers.

`find_max()`
- Finds the largest value.

`find_min()`
- Finds the smallest value.

```python
# Day 2 Assignment
def calculate_total(numbers):
    return sum(numbers)
    
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

def find_max(numbers):
    return max(numbers)

def find_min(numbers):
    return min(numbers)

numbers = [10, 20, 30, 40, 50]

total = calculate_total(numbers)
average = calculate_average(numbers)
maximum = find_max(numbers)
minimum = find_min(numbers)

print("Numbers:", numbers)
print("Total:", total)
print("Average:", average)
print("Maximum:", maximum)
print("Minimum:", minimum)
```

### Exception Handling with Functions
The basic assignment assumes that the function receives a valid
non-empty list.

However, `calculate_average()` requires at least one value because
the average is calculated by dividing the total by the number of
values.

If an empty list is provided, division by zero can occur.

Exception handling can be used to handle this situation safely.

```python
def calculate_average(numbers):
    try:
        return sum(numbers) / len(numbers)
    except ZeroDivisionError:
        return 0


numbers = []

average = calculate_average(numbers)

print("Average:", average)
```

## 14. Grouping Data

Grouping allows us to divide data into groups based on the values
of a particular column and then perform calculations on each group.

For example, if an employee dataset contains a Department column,
we can calculate the average salary for each department.

The basic structure is:

df.groupby("ColumnName")["AnotherColumn"].mean()

For example:

df.groupby("Department")["Salary"].mean()

This means:

1. Group employees by Department.
2. Select Salary.
3. Calculate the average salary for each department.

### Why is Grouping Important?

Grouping is extremely useful in data analysis.

Examples:

- Average salary by department
- Total sales by region
- Average sales by product
- Number of customers by city
- Total revenue by month

```python
data = {
    "Name": ["John", "Mary", "David", "Priya", "Rahul"],
    "Department": ["AI", "Data Science", "AI", "Data Science", "AI"],
    "Salary": [50000, 60000, 55000, 65000, 70000]
}

df_department = pd.DataFrame(data)

print(df_department)
```

```python
avg_salary=df_department.groupby("Department")["Salary"].mean()
print(avg_salary)
```

## 15. Grouping with Multiple Statistics

- Minimum salary

Pandas provides the `agg()` function for applying multiple
aggregation functions.

```python
summary = df_department.groupby("Department")["Salary"].agg(
     ["mean","max","min"]
 )
print(summary)
```

## 14. Bar Chart

### What is a Bar Chart?

A bar chart compares values across different categories.

For our sales dataset, a bar chart can show:

Revenue by Product

For example:

Product → Revenue

Laptop → ₹100,000
Phone → ₹80,000
Tablet → ₹60,000

### Why Use a Bar Chart?

A bar chart is useful for comparing categories.

Examples:

- Revenue by product
- Sales by region
- Quantity by category
- Revenue by department

```python
import matplotlib.pyplot as plt
df.groupby("Product")["Revenue"].sum().plot(kind="bar")
plt.title("Revenue by product")
plt.xlabel("product")
plt.ylabel("revvenue")
plt.show()
```

## 15. Line Chart

### What is a Line Chart?

A line chart is commonly used to show how a value changes over
an ordered sequence, especially over time.

For our sales dataset, we can use:

Date → Revenue

This can help us understand the sales revenue trend over time.

### Why Use a Line Chart?

Line charts are useful for:

- Revenue over time
- Sales over time
- Monthly performance
- Daily sales
- Website traffic
- Temperature trends

The important idea is:

Bar chart → Compare categories

Line chart → Show trends or changes over an ordered sequence

```python
df["Date"] = pd.to_datetime(df["Date"])
daily_revenue = df.groupby("Date")["Revenue"].sum()
plt.plot(daily_revenue.index, daily_revenue.values)
plt.title("Revenue Over Time")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.hist(df["Revenue"], bins=10)
plt.title("Distribution of Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

## 17. Pie Chart

### What is a Pie Chart?

A pie chart shows how different categories contribute to a whole.

For our sales dataset, we can use it to show:

Revenue Share by Product

For example:

Laptop → 40%
Phone → 35%
Tablet → 25%

The complete pie represents 100% of the total revenue.

### Why Use a Pie Chart?

Pie charts are useful when we want to understand the proportion

For example:

- Revenue share by product
- Sales share by region
- Expense share by category

A pie chart should generally be used when there are only a

```python
product_revenue = df.groupby("Product")["Revenue"].sum()
plt.pie(
    product_revenue,
    labels=product_revenue.index,
    autopct="%1.1f%%"
)
plt.title("Revenue Share by Product")
plt.show()
```

```python
product_revenue = df.groupby("Product")["Revenue"].sum()

plt.figure(figsize=(10, 8))

plt.pie(
    product_revenue,
    labels=None,
    startangle=90

)

plt.title("Revenue Share by Product")

plt.legend(
    product_revenue.index,
    title="Products",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)

plt.tight_layout()
plt.show()
```
