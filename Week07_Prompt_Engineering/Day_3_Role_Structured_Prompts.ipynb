## 2. What Is Role Prompting?

Role prompting is a prompting technique where we tell the LLM what role, perspective, or type of assistant it should adopt for the task.

Example:

"You are an experienced data analyst."

Then we provide the task.

Example:

"You are an experienced data analyst.
Analyze the following sales data and identify the top-performing department."

The role provides additional behavioral context.

## 3. Why Use Role Prompting?

Different tasks require different perspectives.

For example:

Data analysis:
"You are an experienced data analyst."

Programming:
"You are a senior Python developer."

Teaching:
"You are a patient AI tutor."

Customer support:
"You are a professional customer-support assistant."

Writing:
"You are a technical writer."

The role helps communicate the desired perspective and behavior.

## 4. Role Prompting Does Not Change the Model

Important:

Role prompting does NOT retrain the LLM.

We are not changing:

- Model parameters
- Model weights
- Training data

We are providing additional instructions in the prompt.

Therefore:
```text
Role Prompt
     ↓
Current Context
     ↓
LLM
     ↓
Response

## 9. What Is a Structured Prompt?

A structured prompt organizes instructions into clearly defined sections.

For example:
```text
Role:
...

Task:
...

Context:
...

Input:
...

Constraints:
...

Output Format:
...
```
This makes the prompt easier to understand, maintain, test, and modify.

## 10. Basic Structured Prompt

A simple structure is:

### ROLE
Who is the model?

### TASK
What should it do?

### CONTEXT
What information does it need?

### INPUT
What should it process?

### CONSTRAINTS
What rules should it follow?

### OUTPUT
How should the answer be returned?

## 11. Example Structured Prompt

ROLE:

You are an experienced data analyst.

TASK:

Analyze the sales information.

CONTEXT:

The data represents monthly company sales.

INPUT:

January: ₹10 lakh

February: ₹12 lakh

March: ₹9 lakh


CONSTRAINTS:

Use only the information provided.

OUTPUT:

Identify the highest-sales month and explain why.

## 12. Why Structure Helps

Compare:

### Unstructured

"Look at these sales numbers and tell me what happened."

### Structured

Role:

You are an experienced data analyst.

Task:

Analyze the sales data.

Input:

January: ₹10 lakh

February: ₹12 lakh

March: ₹9 lakh

Output:

Identify the highest-sales month and explain the result in two sentences.

The second prompt is easier to understand and control.

## 13. Role

The role establishes the desired perspective.

Examples:

"You are an experienced data analyst."

"You are a senior Python developer."

"You are a mathematics tutor."

"You are a customer-support specialist."

"You are a technical writer."

Role = desired perspective or behavior.

## 14. Task

The task tells the model what operation to perform.

Examples:

- Analyze
- Summarize
- Classify
- Translate
- Explain
- Extract
- Compare
- Generate
- Rewrite

Example:

TASK:

## 15. Context

Context provides relevant background information.

Example:

"The customer is asking about a delayed delivery."

This helps the model understand the situation.

Context should be relevant.

Unnecessary context can make the prompt longer without improving the answer.

## 16. Input

Input is the information the model needs to process.

Example:

INPUT:

Customer:
```text
"My order was supposed to arrive yesterday,
but I still haven't received it."
```
The task tells the model what to do.

The input provides the information to process.

## 17. Constraints

Constraints define limitations or rules.

Examples:

- Use simple language.
- Answer in 3 bullet points.
- Do not invent information.
- Use only the provided data.
- Return only the classification label.
- Keep the answer under 100 words.

Constraints make the desired behavior more specific.

## 18. Output Format

Output instructions define how the answer should look.

Examples:

"Return 5 bullet points."

"Return only the label."

"Use a table."

"Answer in two paragraphs."

"Return JSON."

## 29. Prompt Templates

Instead of writing every prompt from scratch, we can create templates.

Example:

ROLE:
[ROLE]

TASK:
[TASK]

CONTEXT:
[CONTEXT]

INPUT:
[INPUT]

CONSTRAINTS:
[CONSTRAINTS]

OUTPUT:
[OUTPUT]

The application can replace the placeholders with actual information.

````python
role = "You are an experienced data analyst."

task = "Analyze the following sales data."

context = "The data represents monthly sales."

data = """
January: ₹10 lakh
February: ₹15 lakh
March: ₹12 lakh
"""

constraints = "Use only the provided data."

output_format = "Return the highest-sales month and sales amount."

prompt = f"""
ROLE:
{role}

TASK:
{task}

CONTEXT:
{context}

INPUT:
{data}

CONSTRAINTS:
{constraints}

OUTPUT:
{output_format}
"""

print(prompt)
````

## 30. Why Use Prompt Templates?

Prompt templates are useful in applications because:

- They are reusable.
- They are easier to maintain.
- They make prompts consistent.
- They allow dynamic inputs.
- They make testing easier.
- They support prompt versioning.

Example:

One template can process thousands of customer messages.

## 34. Conflicting Instructions

Avoid conflicting instructions.

Bad example:

"Give a detailed explanation in one sentence."

These instructions conflict.

Better:

"Give a concise explanation in one sentence."

Another example:

"Use simple language but include highly technical terminology."

If both requirements are necessary, make the intended balance explicit.

## 35. Ambiguous Instructions

Bad:

"Analyze this."

Better:

"Analyze the sales data and identify the top three products by revenue."

The second prompt clearly defines the task.

Good prompt engineering reduces ambiguity.

## 36. Role Prompting Limitations

Role prompting is useful, but it does not guarantee expertise or correctness.

For example:

"You are a world-class doctor."

does not make the model a real doctor.

The role is an instruction about desired behavior or perspective.

The model can still:

- Make mistakes
- Hallucinate
- Misinterpret information
- Lack required context

Role prompting should not replace validation or domain-specific systems.

## 37. Important Principle

A role does not magically add knowledge.

Compare:

"You are an expert financial analyst."

with:

"You are an experienced data analyst.
Analyze only the following provided financial data.
Do not invent missing values."

The second prompt provides useful constraints.

Therefore:
```
Role
+
Relevant Context
+
Clear Task
+
Constraints
=
More controlled behavior

## 38. Structured Prompt Best Practices

1. Clearly define the task.
2. Provide relevant context.
3. Separate input from instructions.
4. Specify important constraints.
5. Define the desired output.
6. Avoid conflicting instructions.
7. Avoid unnecessary information.
8. Use consistent formatting.
9. Use reusable templates.
10. Test prompts with multiple inputs.

# Interview Questions

## Q1. What is role prompting?

Role prompting tells an LLM to adopt a particular role, perspective, or desired behavior.

## Q2. Give an example of role prompting.

"You are an experienced data analyst."

## Q3. Does role prompting retrain the model?

No. It provides instructions in the current prompt and does not normally change the model's learned parameters.

## Q4. What is a structured prompt?

A prompt organized into clear sections such as role, task, context, input, constraints, and output.

## Q5. Why use structured prompts?

They improve clarity, consistency, maintainability, testing, and debugging.

## Q6. What are common components of a structured prompt?

Role, task, context, input, constraints, and output format.

## Q7. What is the difference between role prompting and few-shot prompting?

Role prompting specifies the desired perspective or behavior.

Few-shot prompting provides examples demonstrating the desired task or output pattern.

## Q8. Can role prompting and few-shot prompting be combined?

Yes.

## Q9. Does saying "You are an expert" guarantee correct answers?

No. Role prompting does not guarantee expertise or correctness.

## Q10. Why are constraints useful?

They reduce ambiguity and restrict the model's response to desired behavior or format.

## Q11. What is a prompt template?

A reusable prompt structure containing placeholders that can be filled with dynamic information.

## Q12. Why are prompt templates useful in applications?

They make prompts reusable, consistent, maintainable, and easier to test.

## Q13. What is an ambiguous prompt?

A prompt that does not clearly specify the intended task or requirements.

## Q14. What is a conflicting prompt?

A prompt containing instructions that contradict each other.

## Q15. Does a longer structured prompt always produce better results?

No. Structure should improve clarity, not add unnecessary information.
