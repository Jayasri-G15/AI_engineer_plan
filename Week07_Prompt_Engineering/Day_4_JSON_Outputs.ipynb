## 2. What Is Structured Output?

Structured output means asking an LLM to return information in a predefined format.

Instead of:

"John's review is positive and I think the sentiment
score is around 0.92."

we can request:
```python
{
  "name": "John",
  "sentiment": "positive",
  "score": 0.92
}
```
The second response is easier for software to process.

## 3. Why Do We Need Structured Output?

Humans can understand free-form text easily.

Software prefers predictable structures.

Example free-form response:

"John seems positive about the product, with a score
of about 0.92."

A program would need to extract:
```
Name → John
Sentiment → positive
Score → 0.92
```
With JSON:
```python
{
  "name": "John",
  "sentiment": "positive",
  "score": 0.92
}
```
Python can directly access these fields.

## 26. Convert Python Dictionary to JSON

We can use:

json.dumps()

to convert a Python object into a JSON string.

````python
import json

data = {
    "name": "John",
    "sentiment": "positive",
    "score": 0.92
}

json_text = json.dumps(data)

print(json_text)
````

## 27. Pretty JSON

We can make JSON easier to read using indentation.

````python
print(json.dumps(data, indent=2))
````

## 28. Convert JSON to Python

We can use:

json.loads()

to parse a JSON string into a Python object.

````python
json_text = '''
{
  "name": "John",
  "sentiment": "positive",
  "score": 0.92
}
'''

data = json.loads(json_text)

print(data)
print(data["name"])
print(data["sentiment"])
print(data["score"])
````

## 30. JSON Parsing Errors

Not every text that looks like JSON is valid JSON.

For example:
```
{
  "name": "John",
  "score": 0.92,
}
```
The trailing comma can make the JSON invalid.

LLM applications should validate or safely parse model output.

````python
import json

json_text = '''
{
  "name": "John",
  "sentiment": "positive",
  "score": 0.92
}
'''

try:
    data = json.loads(json_text)
    print("Valid JSON")
    print(data)

except json.JSONDecodeError as error:
    print("Invalid JSON")
    print(error)
````

## 31. JSON Validation

We should check whether the response contains the information our application expects.

For example:

Required fields:

- name
- sentiment
- score

We can verify that these fields exist before using the response.

````python
required_fields = ["name", "sentiment", "score"]

for field in required_fields:
    if field not in data:
        print("Missing field:", field)
````

## 32. Validate Values

We can also validate values.

For example:

sentiment should be:
```
Positive
Negative
Neutral
```
score should be a number.

Validation protects the application from unexpected model output.

````python
allowed_sentiments = {"positive", "negative", "neutral"}

if data["sentiment"] not in allowed_sentiments:
    print("Invalid sentiment")

if not isinstance(data["score"], (int, float)):
    print("Score must be numeric")
````

## 41. Common JSON Mistakes

### Mistake 1

Using single quotes:

{'name': 'John'}

Standard JSON uses double quotes.

Correct:

{"name": "John"}


### Mistake 2

Trailing comma:

{
  "name": "John",
}

Avoid trailing commas.


### Mistake 3

Returning extra text:

Here is your JSON:

{
  "name": "John"
}

If the application expects pure JSON, the extra text can cause parsing problems.


### Mistake 4

Wrong data types:

{
  "score": "0.92"
}

when the application expects a number.


### Mistake 5

Missing fields.

Always validate expected fields.

# Interview Questions

## Q1. What is structured output?

Structured output means returning model-generated information in a predefined format that an application can reliably process.

## Q2. What is JSON?

JSON stands for JavaScript Object Notation and is a lightweight data-interchange format.

## Q3. Why is JSON useful with LLMs?

It provides predictable, machine-readable output that applications can parse and process.

## Q4. What is a JSON key?

A name used to identify a value inside a JSON object.

Example:

"name": "John"

"name" is the key.

## Q5. What is a JSON value?

The data associated with a key.

Example:

"name": "John"

"John" is the value.

## Q6. What data types can JSON contain?

Strings, numbers, booleans, null, arrays, and objects.

## Q7. What is a JSON array?

A collection of values enclosed in square brackets.

Example:

["Python", "Java", "C++"]

## Q8. What is nested JSON?

A JSON structure containing objects or arrays inside another object or array.

## Q9. How do you parse JSON in Python?

Using the json module, commonly with json.loads() for JSON text.

## Q10. What does json.dumps() do?

It converts a Python object into a JSON-formatted string.

## Q11. What does json.loads() do?

It parses a JSON string into a Python object.

## Q12. Why should LLM output be validated?

Because model output can be malformed, incomplete, or contain unexpected values.

## Q13. What is a schema?

A definition of the expected structure, fields, and data types of the output.

## Q14. What is the difference between free-form text and structured output?

Free-form text is flexible and primarily designed for human consumption.

Structured output follows a predictable format that software can process.

## Q15. Does asking for JSON guarantee valid JSON?

Not necessarily. Natural-language instructions alone do not guarantee perfect formatting. Applications should validate the response, and APIs that support structured-output/schema enforcement can provide stronger guarantees.

## Q16. Why are structured outputs useful in production AI applications?

They allow LLM responses to be connected reliably to application logic, databases, APIs, dashboards, and automation.

## Q17. Is JSON only used in AI?

No. JSON is widely used for communication between software systems and APIs.
