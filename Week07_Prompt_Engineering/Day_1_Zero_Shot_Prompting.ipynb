## 1. What Is Prompt Engineering?

Prompt Engineering is the process of designing and improving instructions given to an AI model to obtain a desired output.

A prompt can influence:

- What the model does
- How the model responds
- The format of the response
- The level of detail
- The style of the response
- The information the model focuses on

Simple idea:
```text
Better instruction
      ↓
More useful output

## 2. What Is a Prompt?

A prompt is the input provided to an LLM.

Example:

"What is machine learning?"

This is a prompt.

The model processes the prompt and generates a response.

Flow:
```text
User Prompt
     ↓
    LLM
     ↓
Generated Response

````python
prompt = "What is machine learning?"

print(prompt)
````

## 3. Prompt vs Response

Prompt:

"What is Python?"

Response:

"Python is a high-level programming language..."

Therefore:

Prompt = Input

Response = Output

The prompt provides the task or information that the model should process.

## 4. Why Is Prompt Engineering Important?

LLMs are flexible.

The same model can perform many different tasks.

For example:

- Summarization
- Classification
- Translation
- Question answering
- Code generation
- Information extraction
- Explanation
- Content generation

The prompt tells the model which task we want it to perform.

Therefore:
```text
LLM
+
Prompt
= 
Task-specific behavior

## 5. Prompt Engineering Is Not Model Training

These are different concepts.

### Model Training

Changes the model's learned parameters.

### Prompt Engineering

Changes the instructions given to the model.

Example:

Training:
Model learns from a large dataset.

Prompt Engineering:
We tell the trained model:

"Explain this concept to a beginner."

We are not changing the model.

We are changing the input instructions.

## 6. What Is Zero-Shot Prompting?

Zero-shot prompting means asking an LLM to perform a task without providing examples of the expected task behavior.

The model receives:
```text
Instruction
+
Input
```
and generates:

Output

No examples are provided.

Basic flow:
```text
Prompt
 ↓
LLM
 ↓
Answer

## 7. Why Is It Called Zero-Shot?

"Zero-shot" means:

Number of examples provided = 0

For example:

Prompt:

"Classify this review as Positive or Negative:

'I love this product.'"

We provide:

0 examples

We simply ask the model to perform the classification.

## 8. Zero-Shot Example

Task:

Classify the sentiment.

Prompt:

"Classify the following review as Positive or Negative:

'The product is excellent.'"

The model might respond:

Positive

Notice that we did not provide examples such as:
```text
"I love it" → Positive
"I hate it" → Negative
```
Therefore this is zero-shot prompting.

## 9. Zero-Shot Prompt Structure

A simple zero-shot prompt can contain:
```text
Task
+
Input
```
Example:

Task:
"Translate the following sentence into French."

Input:
"Good morning."

Complete prompt:

"Translate the following sentence into French:

Good morning."

## 10. Zero-Shot Prompt Example — Classification

Prompt:

"Classify the following email as Spam or Not Spam:

'Congratulations! You won a free prize. Click this link now.'"

No examples are provided.

The model must understand the task from the instruction itself.

## 11. Zero-Shot Prompt Example — Summarization

Prompt:

"Summarize the following paragraph in two sentences:

Artificial intelligence is increasingly being used in healthcare,
finance, education, transportation, and customer service."

The model performs summarization without being shown an example.

## 12. Zero-Shot Prompt Example — Translation

Prompt:

"Translate this sentence into Tamil:

Artificial intelligence is changing the world."

The model receives the instruction and input directly.

No translation examples are provided.

## 13. Zero-Shot Prompt Example — Question Answering

Prompt:

"What is the capital of France?"

The model receives only the question.

No examples are provided.

Therefore this is also a zero-shot task.

## 14. Clear Prompt vs Unclear Prompt

Consider:

Prompt A:

"Tell me about Python."

Prompt B:

"Explain Python to a beginner in 5 simple bullet points. Include one small example."

Prompt B gives the model more useful instructions.

It specifies:

- Topic
- Audience
- Length
- Format
- Example requirement

The model has less ambiguity about what we want.

## 15. Prompt Specificity

Prompt specificity means clearly defining what the model should do.

Less specific:

"Explain databases."

More specific:

"Explain relational databases to a beginner using a simple real-world example in 5 bullet points."

The second prompt provides:

- Task
- Audience
- Explanation level
- Example requirement
- Output structure

More specific instructions often make the output more predictable.

## 16. Task Instruction

The task instruction tells the model what operation to perform.

Examples:

"Summarize this text."

"Translate this sentence."

"Classify this review."

"Extract the person's name."

"Explain this concept."

"Generate Python code."

The task should be clear.

## 17. Context

Context provides additional information that helps the model understand the task.

Example:

Without context:

"Explain it."

What is "it"?

The model may not know.

With context:

"Explain overfitting in machine learning to a beginner."

Now the model knows the topic and intended audience.

## 18. Output Expectations

We can tell the model how we want the response structured.

Example:

"Explain APIs in 5 bullet points."

Expected structure:

- Point 1
- Point 2
- Point 3
- Point 4
- Point 5

Another example:

"Answer in one sentence."

Output expectation makes the desired response format clearer.

## 19. Prompt Components

A useful prompt can contain:

1. Task
2. Context
3. Input
4. Constraints
5. Output format

Example:

Task:
"Summarize"

Context:
"This is a technical article."

Input:
Article text

Constraint:
"Use simple language."

Output format:
"5 bullet points"

Not every prompt needs every component.

The important idea is to provide the information necessary for the task.

## 20. Basic Prompt Template

A useful general structure is:

Task:
[What should the model do?]

Context:
[What information does the model need?]

Input:
[What should the model process?]

Constraints:
[What limitations should be followed?]

Output:
[How should the answer be formatted?]

## 21. Example Prompt Template

Task:
"Classify the sentiment of the review."

Context:
"The review is about a mobile phone."

Input:
"The battery life is excellent, but the camera is poor."

Constraints:
"Use only Positive, Negative, or Neutral."

Output:
"Return one label."

This is still zero-shot.

Why?

Because we provided:

0 examples.

We only provided instructions.

## 25. Zero-Shot Prompt with Gemini

If you are continuing from Week 6, you can use your existing Gemini client.

The important part is the prompt.

We will create a prompt and send it to the model.

## 26. Analyze the Prompt

Our prompt contains:

### Task
Classify the review.

### Possible outputs
Positive, Negative, or Neutral.

### Input
The customer review.

### Output constraint
Return only one label.

This is a zero-shot prompt because we did not provide any examples.

## 27. Prompt Improvement

Suppose we start with:

"Classify this review."

This is vague.

We can improve it:

"Classify the following customer review as Positive, Negative, or Neutral."

Then improve further:

"Classify the following customer review as Positive, Negative, or Neutral.

Return only one label and do not provide an explanation.

Review:
'The product arrived late but the quality is excellent.'"

Each improvement reduces ambiguity.

## 28. Prompt Versioning

When developing prompts, it is useful to create versions.

Example:

Prompt V1:
"Summarize this text."

Prompt V2:
"Summarize this text in 5 bullet points."

Prompt V3:
"Summarize this technical text in 5 bullet points using beginner-friendly language."

We can compare the outputs.

This is the beginning of prompt evaluation and optimization.

## 29. Prompt Evaluation

Prompt evaluation means checking whether a prompt produces useful and reliable outputs.

We can evaluate:

- Correctness
- Relevance
- Consistency
- Format
- Completeness
- Clarity

Example:

Prompt A → unclear output

Prompt B → consistent output

We would prefer Prompt B.

Later in Week 7 we will study systematic prompt evaluation in more detail.

## 30. Prompt Optimization

Prompt optimization means improving a prompt to obtain better results.

Example:

V1:
"Explain AI."

V2:
"Explain AI to a beginner."

V3:
"Explain AI to a beginner using a simple real-world example in 5 bullet points."

We gradually improve the instructions.

Optimization does NOT necessarily mean making the prompt longer.

The goal is:
```
Clearer instruction
+
Relevant context
+
Useful constraints
=
Better output

## 31. Longer Prompt Does Not Always Mean Better Prompt

Bad assumption:

Longer prompt = Better prompt

This is not necessarily true.

A prompt should contain useful information.

Avoid:

- Unnecessary instructions
- Repeated instructions
- Conflicting instructions
- Irrelevant information

Good prompt engineering focuses on clarity and relevance.

## 32. Common Zero-Shot Prompting Mistakes

### Mistake 1: Vague instruction

"Tell me something about AI."

### Mistake 2: Missing context

"Explain this."

### Mistake 3: No output expectation

The model may produce a long response when we wanted one sentence.

### Mistake 4: Conflicting instructions

"Be extremely detailed but answer in one sentence."

### Mistake 5: Unnecessary information

Adding irrelevant context can make the task less clear.

## 33. Zero-Shot Prompting Limitations

Zero-shot prompting can work very well for many common tasks.

However, it can struggle when:

- The task is unusual
- The desired output is highly specific
- The instructions are ambiguous
- The model lacks necessary context
- The task requires a particular style or pattern

In such cases, other techniques can help.

Examples:

- Few-shot prompting
- Role prompting
- Structured prompts
- Structured outputs
- Prompt optimization

These will be covered later in Week 7.
