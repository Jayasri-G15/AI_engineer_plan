## 2. What Is Few-Shot Prompting?

Few-shot prompting means providing a small number of examples inside the prompt before asking the model to perform the actual task.

Example:

Example 1:
"I love this phone." → Positive

Example 2:
"This phone is terrible." → Negative

Now classify:

"The phone is excellent." → ?

The examples demonstrate the expected task pattern.

## 3. Why Use Examples?

Instructions sometimes leave room for interpretation.

For example:

"Classify the text."

The model may not know:

- What labels to use
- How the labels should be written
- What output format is expected
- How the task should be interpreted

Examples provide additional guidance.

Instead of only saying:

"Do this task."

we show:

"Here are examples of how this task should be performed."

## 4. Basic Few-Shot Flow

The basic process is:
```text
Example 1
     ↓
Example 2
     ↓
Example 3
     ↓
Instruction
     ↓
New Input
     ↓
    LLM
     ↓
  Output
```
The examples and instruction are included in the prompt.

## 5. What Does "Few" Mean?

Few-shot does not mean a specific fixed number.

It means using a small number of examples.

For example:

1 example → one-shot prompting

2 examples → few-shot prompting

3 examples → few-shot prompting

5 examples → few-shot prompting

The exact number depends on:

- Task complexity
- Available context
- Example quality
- Desired behavior

## 18. Examples Teach Patterns

The important idea is:
```text
Examples
     ↓
Demonstrate Pattern
     ↓
LLM
     ↓
Apply Pattern to New Input
```
The model is not being permanently trained by these examples.

The examples are provided as part of the current prompt.

## 19. Few-Shot Prompting Is Not Training

This distinction is extremely important.

### Training

Changes the model's learned parameters.

### Few-Shot Prompting

Provides temporary examples in the prompt.

The model uses the examples while processing the current request.

Therefore:

Few-shot prompting ≠ model training

## 20. In-Context Learning

Few-shot prompting is closely related to a concept called:

In-Context Learning

The model receives examples inside the context and uses them to determine how to perform the task.

Conceptually:
```text
Examples
   ↓
Context
   ↓
LLM
   ↓
Pattern
   ↓
New Input
   ↓
Output

## 21. Important Difference

The model is not updating its parameters when you provide examples in a normal prompt.

Instead:
```text
Prompt examples
     ↓
Temporary context
     ↓
Model processes context
     ↓
Generates response
```
After the request is finished, those examples do not permanently retrain the model.

## 23. Bad Few-Shot Examples

Poor examples can confuse the model.

Problems include:

- Incorrect labels
- Ambiguous examples
- Inconsistent formatting
- Irrelevant examples
- Conflicting examples

Example:

Input:
"I love this product."

Output:
Negative

This is a poor example because the demonstrated label conflicts with the likely intended sentiment.

Bad examples can produce bad behavior.

We should not randomly add examples.

Good examples should represent the task.

```text

## 27. Too Many Examples

More examples are not automatically better.

More examples can:

- Increase prompt size
- Consume more tokens
- Increase cost for paid APIs
- Reduce available context
- Introduce irrelevant information
- Introduce contradictions

The goal is:

Small number of high-quality examples

rather than:

Large number of random examples.

## 33. Dynamic Few-Shot Prompts

```
The application can construct a prompt using these examples.

Flow:
```text
Stored Examples
      ↓
Prompt Builder
      ↓
Few-Shot Prompt
      ↓
LLM
      ↓

## 34. Few-Shot Prompt Builder

Python can construct the prompt automatically.

Example:
```text
examples
    ↓
loop through examples
    ↓
create formatted examples
    ↓
add new input
    ↓
send prompt to LLM
```
This becomes useful when building real AI applications.

````python
examples = [
    ("I love this product.", "Positive"),
    ("This product is terrible.", "Negative"),
    ("The product arrived yesterday.", "Neutral")
]

new_review = "The product quality is excellent."

prompt = "Classify customer reviews.\n\n"

for i, (text, label) in enumerate(examples, start=1):
    prompt += f"Example {i}:\n"
    prompt += f"Review: {text}\n"
    prompt += f"Label: {label}\n\n"

prompt += f"Now classify:\nReview: {new_review}\n"
prompt += "Return only the label."

print(prompt)
````

## 36. Few-Shot Prompting Limitations

Few-shot prompting is powerful, but it has limitations.

### 1. Context usage

Examples consume context.

### 2. Cost

For paid APIs, more input tokens can increase cost.

### 3. Bad examples

Incorrect examples can lead to poor outputs.

### 4. Too many examples

Unnecessary examples can make prompts inefficient.

### 5. Conflicting examples

The model may receive contradictory patterns.

### 6. Task complexity

Some tasks require more than examples.

The examples must therefore be selected carefully.

## 37. Few-Shot Prompting Best Practices

1. Keep examples relevant.
2. Use correct examples.
3. Keep formatting consistent.
4. Cover important classes or patterns.
5. Avoid unnecessary examples.
6. Make instructions clear.
7. Separate examples from the actual input.
8. Test multiple examples.
9. Evaluate the output.
10. Optimize the examples when necessary.
