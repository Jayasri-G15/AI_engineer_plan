# 1. Why Do Agents Need Memory?

Consider a conversation:

User:
"My name is Jayasri."

Agent:
"Nice to meet you."

Later:

User:
"What is my name?"

If the system has no access to previous information,
it may not know the answer.

Memory allows an agent system to retain useful information.

Memory can be used for:

- Previous conversation
- User preferences
- Previous tool results
- Task progress
- Important facts
- Previous interactions

# Example

User:
"I prefer Python for data analysis."

Later:

User:
"Create a data-analysis example."

A memory-enabled system may retrieve:

User preference:
Python

The agent can then use Python in the response.

The important idea is:

Memory provides information that can influence future actions.

# 2. Context vs Memory

These concepts are related but different.

## Context

Information currently available to the model.

Example:

The current conversation messages.

## Memory

Information intentionally retained so that it can be used later.

Example:

A stored preference:

{
    "preferred_language": "Python"
}

### Simple distinction

Context:
"What information is available right now?"

Memory:
"What information has the system decided to retain for future use?"


# Important

Not everything in a conversation should automatically become
long-term memory.

A memory system should consider:

- Relevance
- Usefulness
- Persistence
- Privacy
- Accuracy
- Expiration

Good memory design is selective.

More stored information does not automatically mean a better agent.

Several memory concepts are useful in agent systems.

## 2. Long-Term Memory

# 4. Short-Term Memory

Short-term memory contains information needed during the
current interaction.

Example:

User:
"Calculate the total of 100 and 200."

Agent:
"300."

The current task information may be represented as:

{
    "numbers": [100, 200],
    "operation": "addition",
    "result": 300
}

Once the task ends, this information may no longer need to persist.

````python
short_term_memory = {
    "current_task": "Calculate total sales",
    "current_step": "calculation",
    "data_source": "sales.csv",
    "partial_result": 125000
}

print(short_term_memory)
````

# 5. Long-Term Memory

Long-term memory stores information that may be useful in
future interactions.

Example:

User preference:

{
    "preferred_format": "PDF",
    "preferred_language": "Python"
}

Later, the agent can retrieve these preferences.

Long-term memory may be stored in:

- PostgreSQL
- Redis
- Document databases
- Vector databases
- Key-value stores
- Specialized memory systems

The storage choice depends on the type and scale of information.

````python
long_term_memory = {
    "preferred_language": "Python",
    "preferred_report_format": "PDF",
    "favorite_dashboard_tool": "Power BI"
}

print(long_term_memory)
````

````python
preference = long_term_memory.get(
    "preferred_language"
)

print("Preferred language:", preference)
````

# 6. Working Memory

Working memory stores temporary information required while
the agent is solving a task.

Example:

Task:
"Calculate the average order value."

The agent may temporarily store:

total_sales = 500000
total_orders = 1000

Then:

average_order_value = 500

This information is useful during the task.

It may not need to become permanent memory.

````python
working_memory = {
    "total_sales": 500000,
    "total_orders": 1000
}

working_memory["average_order_value"] = (
    working_memory["total_sales"]
    / working_memory["total_orders"]
)

print(working_memory)
````

# 8. Simple Memory Retrieval

For learning purposes, we can create a basic keyword-based
memory retrieval system.

This is NOT a production memory system.

It is only to understand the concept.

If the current request contains "sales",
we can retrieve memories related to sales.

````python
def retrieve_memory(query, memories):
    query_words = set(
        query.lower().split()
    )

    relevant = []

    for memory in memories:
        memory_words = set(
            memory.lower().split()
        )

        if query_words.intersection(memory_words):
            relevant.append(memory)

    return relevant
````

````python
query = "Create a sales analysis workflow."

relevant_memories = retrieve_memory(
    query,
    memories
)

print("Relevant memories:")

for memory in relevant_memories:
    print("-", memory)
````

# 13. What Is Agent Planning?

Planning means deciding what actions should be performed
to accomplish a goal.

Example:

Goal:

"Analyze sales data and prepare a report."

Possible plan:

1. Locate dataset
2. Load dataset
3. Validate data
4. Clean data
5. Calculate KPIs
6. Identify trends
7. Generate report
8. Return result

Planning converts:

High-level goal

into:

Actionable steps.

# 14. Goal Decomposition

Complex goals can be broken into smaller tasks.

Example:

Goal:
"Create a market analysis report."

Break it into:

1. Collect market data
2. Clean data
3. Calculate indicators
4. Analyze trends
5. Compare assets
6. Generate charts
7. Write report

This is called task decomposition.

The smaller tasks can then be assigned to tools or specialized agents.

````python
goal = "Analyze sales data and create a report"

plan = [
    "Load sales data",
    "Validate data",
    "Clean data",
    "Calculate sales KPIs",
    "Identify trends",
    "Generate report"
]

print("Goal:")
print(goal)

print("\nPlan:")

for step_number, step in enumerate(plan, start=1):
    print(f"{step_number}. {step}")
````

# 15. Sequential Planning

Some tasks must happen in a specific order.

Example:

You cannot calculate average order value
before loading the sales data.

Therefore:

Load Data
↓
Validate Data
↓
Calculate Metrics
↓
Generate Report

The order matters.

This is sequential planning.

# 16. Task Dependencies

Tasks can depend on previous tasks.

Example:

Task A:
Load dataset.

Task B:
Clean dataset.

Task C:
Calculate KPIs.

Task D:
Write report.

Dependencies:

A → B → C → D

If Task A fails:

B cannot continue.

If Task C fails:

D may not have the required information.

Understanding dependencies is important for reliable planning.

Parallel execution can reduce overall latency.

# 18. Dynamic Planning

A plan does not always need to be fixed.

An agent may change its plan after observing results.

Example:

Initial plan:

1. Read sales file
2. Calculate KPIs
3. Create report

After reading the file, the agent discovers:

"The file has missing dates."

The agent may change the plan:

1. Read sales file
2. Detect missing dates
3. Clean date values
4. Validate data
5. Calculate KPIs
6. Create report

This is dynamic planning.

# 20. Planning + Tools

Planning determines:

"What should I do?"

Tool calling determines:

"How can I perform that action?"

Example:

Goal:
"Find total sales."

Plan:
1. Read sales data.
2. Calculate total.

Tools:

Step 1:
File Reader

Step 2:
Python Calculator

Therefore:

Planning
↓
Tool Selection
↓
Tool Execution
↓
Observation
↓
Next Planning Step

````python
agent_state = {
    "goal": "Find total sales",
    "plan": [
        "Read sales data",
        "Calculate total sales",
        "Return result"
    ],
    "current_step": 0,
    "observations": []
}

print(agent_state)
````

````python
# Complete the first step

agent_state["observations"].append(
    "Sales data loaded successfully."
)

agent_state["current_step"] = 1

print(agent_state)
````

````python
# Complete the second step

agent_state["observations"].append(
    "Total sales = ₹50,00,000."
)

agent_state["current_step"] = 2

print(agent_state)
````

# 21. Planning Loop

A simplified planning loop looks like:

while task_not_complete:

    understand_current_state()

    decide_next_action()

    execute_action()

    observe_result()

    update_state()

    if necessary:
        replan()

return final_result

The exact implementation differs across agent frameworks.

````python
steps = [
    "Load data",
    "Validate data",
    "Calculate KPIs",
    "Generate report"
]

current_step = 0

while current_step < len(steps):
    print(
        f"Executing: {steps[current_step]}"
    )

    current_step += 1

print("Task completed.")
````

# 22. What Happens When a Plan Fails?

Suppose:

Plan:

1. Read sales.csv
2. Calculate total sales
3. Generate report

But:

sales.csv does not exist.

The agent observes:

"File not found."

It could:

- Search for another file
- Ask the user for the correct file
- Try another data source
- Stop and report the problem

The correct behavior depends on the system design.

The important idea is:

Observation can change the next action.

# 23. Retry vs Re-Plan

These are not the same.

## Retry

Repeat the same action.

Example:

API call failed due to a temporary network error.

Retry the API call.

## Re-Plan

Choose a different strategy.

Example:

Database tool fails permanently.

Alternative:

Use a cached dataset.

Therefore:

Temporary failure
→ Retry

Strategic failure
→ Re-plan

# 24. Planning Constraints

An agent should not have unlimited freedom.

Useful constraints include:

- Maximum number of steps
- Maximum tool calls
- Time limit
- Budget limit
- Allowed tools
- Required approval
- Allowed data sources

Example:

Maximum tool calls = 10

If the agent reaches 10 calls:

Stop execution
and return an appropriate status.

Constraints help prevent uncontrolled loops and unnecessary cost.

# 27. Memory vs RAG

Memory and RAG can look similar because both involve retrieving
information.

But they solve different problems.

## RAG

Usually retrieves information from an external knowledge base.

Example:

Company handbook

## Memory

Usually stores information about previous interactions, user
preferences, task state, or experiences.

Example:

User preference

→ Retrieve preference

Both can use vector databases, but their purpose and lifecycle
can be different.

# 29. Planning in a Multi-Agent System

Our Week 9 project will contain:

Main Agent
├── Research Agent
├── Data Agent
└── Writer Agent

The Main Agent can plan:

1. Research the topic.
2. Analyze the provided data.
3. Give research + data results to Writer Agent.
4. Ask Writer Agent to create final report.

Each specialized agent may also have its own internal plan.

Therefore:

Main Plan
+
Agent-Specific Plans
=
Multi-Agent Workflow

````python
multi_agent_state = {
    "goal": "Create a market research report",

    "research_agent": {
        "status": "pending",
        "result": None
    },

    "data_agent": {
        "status": "pending",
        "result": None
    },

    "writer_agent": {
        "status": "pending",
        "result": None
    },

    "final_status": "in_progress"
}

print(multi_agent_state)
````

````python
multi_agent_state["research_agent"] = {
    "status": "completed",
    "result": "Research findings collected."
}

multi_agent_state["data_agent"] = {
    "status": "completed",
    "result": "Sales data analyzed."
}

print(multi_agent_state)
````

# Observation

The main agent can inspect the state and determine:

Research Agent → completed
Data Agent → completed
Writer Agent → pending

Therefore, the next action can be:

Send research + data results to Writer Agent.

This is how shared state supports coordination.

# 30. Memory Design Principles

A good agent memory system should consider:

## Relevance

Store information that may actually be useful.

## Accuracy

Incorrect memory can cause incorrect future behavior.

## Freshness

Some information becomes outdated.

## Privacy

Sensitive information requires careful handling.

## Retrieval

Stored information must be easy to retrieve when needed.

## Expiration

Some temporary memories should eventually disappear.

## Scope

A memory may belong to:

- Current task
- User
- Organization
- Agent
- Session

Memory should have a clear purpose.

# 31. Planning Design Principles

A useful agent plan should be:

### Clear

Each step should have a meaningful objective.

### Achievable

The agent should have the required tools.

### Ordered

Dependencies should be respected.

### Observable

The agent should be able to inspect results.

### Adaptable

The plan should be able to change when conditions change.

### Bounded

The system should limit excessive steps and tool calls.

Good planning is not about creating the longest possible plan.

It is about creating the necessary plan.

# 33. Interview Questions and Answers

### Q1. What is agent memory?

Agent memory is information retained by an agent system so that
it can be used during the current task or future interactions.

---

### Q2. What is the difference between context and memory?

Context is information currently available to the model.

Memory is information intentionally stored for later retrieval
and use.

---

### Q3. What is short-term memory?

Short-term memory contains information needed during the current
conversation or task.

---

### Q4. What is long-term memory?

Long-term memory stores information that can persist across
interactions.

---

### Q5. What is working memory?

Working memory contains temporary information required while
solving the current task.

---

### Q6. What is agent state?

Agent state represents the current status of a task, including

---

### Q7. What is agent planning?

Planning is the process of determining the actions required

---

### Q8. What is task decomposition?

Task decomposition means breaking a complex goal into smaller

---

### Q9. What is sequential planning?

Sequential planning organizes tasks in an order where later

---

### Q10. What is dynamic planning?

Dynamic planning allows the agent to modify its plan based on

---

### Q11. What is re-planning?

Re-planning means creating or modifying the current plan after

---

### Q12. What is the difference between retry and re-planning?

Retry repeats the same action, usually after a temporary failure.

Re-planning changes the strategy or chooses a different action.

---

State allows the agent to track what has already happened,
what remains, and what information is available.

---

latency, loops, or unintended actions.

---

### Q16. What is multi-agent planning?

Multi-agent planning coordinates tasks among multiple specialized
agents so that their outputs contribute to a larger goal.
