# 1. What Is a Multi-Agent System?

A multi-agent system contains multiple AI agents that collaborate to complete a larger task.

Instead of one agent doing everything:

User
↓
Single Agent
↓
Final Answer

we can divide the problem:

User
↓
Main Agent
↓
Specialized Agents
↓
Results
↓
Final Answer

Each agent can have a specific responsibility.

Example:

Research Agent
Data Analyst Agent
Writer Agent
Reviewer Agent

Together they solve a larger problem.

# 2. Single-Agent vs Multi-Agent

## Single-Agent System

One agent handles the entire task.

User
 ↓
Agent
 ├── Search
 ├── Analyze
 ├── Calculate
 ├── Write
 └── Review
 ↓
Answer

## Multi-Agent System

Different agents specialize in different responsibilities.

User
 ↓
Manager Agent
 ├── Research Agent
 ├── Data Agent
 ├── Writer Agent
 └── Reviewer Agent
 ↓
Final Answer

# 3. Why Use Multiple Agents?

A large task can contain multiple types of work.

For example:

"Create a market research report."

The task may require:

1. Web research
2. Data collection
3. Data analysis
4. Competitor analysis
5. Report writing
6. Fact checking

Instead of one agent handling everything, we can assign specialized responsibilities.

### Benefits

- Separation of responsibilities
- Specialized prompts
- Specialized tools
- Easier debugging
- Parallel execution
- Better workflow control
- Easier scaling
- Clearer architecture

# 4. Important Principle: Specialization

A multi-agent system works best when agents have clear responsibilities.

Example:

### Research Agent

Responsible for finding information.

### Data Agent

Responsible for processing numerical data.

### Writer Agent

Responsible for producing the final report.

### Reviewer Agent

Responsible for checking the output.

Each agent should have a clear job.

Avoid creating multiple agents that all perform the same task without a reason.

- What information to provide

- Whether the task is complete

# 17. Supervisor Decision Example

User:

"Build a sales analysis report."

Supervisor:

1. Send data task to Data Agent.
2. Send market research to Research Agent.
3. Wait for results.
4. Send combined information to Analyst Agent.
5. Send report to Reviewer Agent.
6. If reviewer finds issues, send back for correction.
7. Return final report.

The supervisor controls the workflow.

# 20. Retry vs Re-plan in Multi-Agent Systems

These concepts are different.

## Retry

The same agent tries the same operation again.

Example:

API timeout
↓
Retry API request

## Re-plan

The system changes its strategy.

Example:

Search tool failed
↓
Supervisor chooses database tool
↓
Continue workflow

Retry:
"Try again."

Re-plan:
"Try something different."

# 31. Simple Agent Pipeline in Python

We can simulate a multi-agent system without using a framework.

Each function represents an agent.

```python
def data_agent(state):
    state["data_result"] = "Data loaded and validated"
    return state

def analysis_agent(state):
    state["analysis_result"] = "Sales increased by 12%"
    return state

def writer_agent(state):
    state["draft"] = "Quarterly sales increased by 12%."
    return state

Then:

state = data_agent(state)
state = analysis_agent(state)
state = writer_agent(state)

This demonstrates the architecture before introducing a framework.

````python
def data_agent(state):
    state["data_result"] = "Data loaded and validated"
    return state


def analysis_agent(state):
    state["analysis_result"] = "Sales increased by 12%"
    return state


def writer_agent(state):
    state["draft"] = "Quarterly sales increased by 12%."
    return state


state = {
    "data_result": None,
    "analysis_result": None,
    "draft": None
}

state = data_agent(state)
state = analysis_agent(state)
state = writer_agent(state)

print(state)
````

# 32. Conditional Routing

A multi-agent workflow may need to choose different paths.

Example:

Supervisor
     |
     v
Check Request
   /     \
  /       \
Data     Research
  |         |
  +----+----+
       |
       v
     Writer

The routing decision can depend on:

- User request
- Agent output
- Tool result
- Validation result
- State
- Business rules

````python
def route_task(task_type):
    if task_type == "data":
        return "Data Agent"
    elif task_type == "research":
        return "Research Agent"
    elif task_type == "writing":
        return "Writer Agent"
    else:
        return "Unknown"


print(route_task("data"))
print(route_task("research"))
````

# 35. Infinite Loop Problem

A multi-agent system can accidentally continue forever.

Example:

Agent A
 ↓
Agent B
 ↓
Reviewer
 ↓
Agent A
 ↓
Agent B
 ↓
Reviewer
 ↓
...

The system needs a stop condition.

Examples:

- Maximum number of iterations
- Maximum time
- Maximum cost
- Maximum tool calls
- Reviewer approval
- Confidence threshold
- Explicit completion state

Security becomes more important as the number of agents increases.

### Authentication
Who can access the system?

### Authorization

### Input Validation

# 40. When Should You Use Multi-Agent Architecture?

Multi-agent architecture can be useful when:

- The task contains clearly different responsibilities.
- Different agents need different tools.
- Tasks can run independently.
- Different expertise or prompts are useful.
- The workflow needs specialized reviewers.
- Complex coordination is required.
- Multiple perspectives are valuable.

A single agent may be better when:

- The task is simple.
- There are few tools.
- The workflow is predictable.
- Multi-agent coordination would add unnecessary complexity.

# 49. Interview Questions

## Q1. What is a multi-agent system?

A multi-agent system consists of multiple AI agents that collaborate or coordinate to solve a larger task.

---

## Q2. Why use multiple agents?

Multiple agents allow a complex task to be divided into specialized responsibilities.

---

## Q3. What is agent specialization?

Assigning an agent a focused role, goal and set of tools for a specific responsibility.

---

## Q4. What is a supervisor agent?

A supervisor agent coordinates specialized agents and decides which agent should perform particular tasks.

---

## Q5. What is shared state?

Shared state is information maintained by the workflow and made available to relevant agents during task execution.

# 50. Interview Questions - Continued

## Q6. What is agent handoff?

A handoff occurs when responsibility for a task is transferred from one agent to another.

---

## Q7. What is sequential multi-agent architecture?

Agents execute one after another, where the output of one stage can become the input of the next.

---

## Q8. What is parallel multi-agent architecture?

Independent agents execute separate tasks simultaneously and their results are later combined.

---

## Q9. What is hierarchical architecture?

A hierarchical system organizes agents into multiple levels, where higher-level agents coordinate lower-level agents.

---

## Q10. What is human-in-the-loop?

A human participates in or approves specific agent decisions or actions.

---

## Q11. Why is least privilege important?

Each agent should receive only the permissions and tools necessary for its responsibility, reducing security and operational risk.

# 51. Interview Questions - Advanced

## Q12. What problems can occur in multi-agent systems?

Possible problems include:

- Agent loops
- Conflicting outputs
- Tool failures
- Incorrect handoffs
- State corruption
- High latency
- High cost
- Excessive model calls
- Security problems

---

## Q13. How do you prevent infinite agent loops?

Use controls such as:

- Maximum iterations
- Maximum execution time
- Maximum tool calls
- Explicit completion conditions
- Human approval
- Error states

---

## Q14. How do you choose between single-agent and multi-agent architecture?

First analyze the problem.

If responsibilities are simple and closely related, a single agent may be sufficient.

If the problem contains clearly separable specialized tasks, multiple agents may provide architectural benefits.

---

## Q15. How can LangGraph help with multi-agent systems?

LangGraph can represent agents as nodes in a stateful graph and define transitions between them using edges and conditional routing.
