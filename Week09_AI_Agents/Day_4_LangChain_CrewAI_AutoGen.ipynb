# 9. Chains vs Agents

This is an important interview concept.

## Chain

A chain follows a predefined sequence.

Example:

Input
↓
Clean data
↓
Calculate KPI
↓
Generate report

The developer defines the sequence.

## Agent

An agent can decide what action to take.

Example:

User asks:

"Analyze this sales dataset."

The agent may decide:

1. Load the dataset
2. Inspect columns
3. Clean missing values
4. Calculate KPIs
5. Create visualizations
6. Generate summary

The exact actions can depend on the task.

# 10. Chain Example

Suppose we want to process customer feedback.

Fixed workflow:

Feedback
↓

↓

↓

This is a chain because the steps are predefined.

Example:

    sentiment = classify_sentiment(cleaned)
    summary = generate_summary(sentiment)

    return summary

A chain is useful when the workflow is predictable.

# 34. Simple Agent Without a Framework

A basic agent loop can be represented with Python:

while not finished:

    decision = llm_decide(state)

    if decision == "search":
        result = search_tool()
        state.append(result)

    elif decision == "calculate":
        result = calculator_tool()
        state.append(result)

    elif decision == "finish":
        break

This demonstrates that an agent framework is an abstraction over patterns like these.

````python
tools = {
    "add": lambda a, b: a + b,
    "multiply": lambda a, b: a * b
}

def execute_tool(tool_name, a, b):
    if tool_name not in tools:
        raise ValueError("Unknown tool")

    return tools[tool_name](a, b)

print(execute_tool("add", 10, 20))
print(execute_tool("multiply", 10, 20))
````

# 56. What is LangGraph?

LangGraph is a framework/library for building **stateful, multi-step AI agent workflows**.

It is designed around the idea of representing an AI application as a **graph**.

A graph contains:

- Nodes
- Edges
- State

### Node

A node represents an action or step.

Examples:

- Call an LLM
- Search documents
- Call a tool
- Validate an answer
- Ask for human approval

### Edge

An edge defines what happens next.

It can be:

- A fixed transition
- A conditional transition

### State

State contains the information that moves through the workflow.

For example:

- User question
- Tool results
- Conversation history
- Retrieved documents
- Current task status
- Agent decisions

---

## Simple LangGraph Mental Model

```text
                START
                  |
                  v
             Understand
                  |
                  v
             Choose Tool
               /     \
              /       \
             v         v
         Search     Calculator
             \         /
              \       /
               v     v
              Review
                 |
          +------+------+
          |             |
       Complete       Retry
          |             |
          v             |
         END <----------+

The graph controls how the agent moves between steps.

LangGraph vs LangChain

A useful beginner-level distinction:

LangChain

Provides components and abstractions for building LLM applications.

Think:

Models + Prompts + Tools + Retrievers + Agents + Workflows

LangGraph

Focuses on building stateful, graph-based agent workflows where the application can move through nodes, make decisions, loop, pause, resume and maintain state.

Think:

State + Nodes + Edges + Control Flow

LangGraph can work with LangChain components.

Why LangGraph Is Useful

LangGraph is particularly useful when an agent needs:

Multiple steps
Conditional branching
Loops
State management
Tool calls
Human approval
Retry logic
Re-planning
Multiple agents
Long-running workflows

Example:

User
↓
Agent
↓

↓
Check Result
↓
Enough Information?
├── No → Search Again
└── Yes
↓
Generate Answer
↓
END

This type of controlled workflow is one of the main ideas behind LangGraph.

Key Mental Model

Remember:

LangGraph → Stateful graph-based agent workflows

LangGraph is especially useful when the agent's workflow needs explicit control over state, steps, branching and loops.
