## What is a Large Language Model?

A Large Language Model (LLM) is a deep learning model trained on very large amounts of text data to learn patterns in language.

LLMs can perform tasks such as:

- Generating text
- Answering questions
- Summarizing text
- Translating languages
- Writing code
- Extracting information
- Classifying text
- Following instructions

Examples of applications include:

- AI chatbots
- Coding assistants
- Document assistants
- Customer support systems
- Writing assistants

**An LLM is a large neural-network model trained on lots of text to learn patterns in language and generate useful responses.**

## Why Are They Called "Large" Language Models?

The word "Large" generally refers to the scale of the model and its training.

Large language models can involve:

- Very large datasets
- Very large neural networks
- Large numbers of parameters
- Significant computational resources

The exact size varies from model to model.

"Large" does not simply mean that the model contains a large text file.

It refers mainly to the scale of the learned model and its training. 

## What Does an LLM Actually Learn?

An LLM does not simply memorize a dictionary.

During training, it learns statistical patterns and relationships in language.

For example, after seeing many examples, a model can learn relationships between:

- Words
- Phrases
- Sentences
- Concepts
- Context
- Syntax
- Different styles of language

Example:

"The sky is usually ___."

A language model may learn that:

"blue"

is a highly likely continuation.

The model learns these patterns from its training process.     

## What is LLM Training?

Training is the process through which a model learns patterns from large amounts of data.

A simplified training process is:
```text 
Large Text Dataset
       ↓
Tokenization
       ↓
Neural Network
       ↓
Prediction
       ↓
Compare Prediction with Target
       ↓
Calculate Error
       ↓
Update Model Parameters
       ↓
Repeat Many Times
```
The model gradually adjusts its parameters so that its predictions improve.

## Simple Example of Language Model Training

Suppose the training text contains:

"The cat sat on the mat."

The model may receive:

"The cat sat on the"

and try to predict:

"mat"

If the prediction is incorrect:

Prediction:
"house"

Target:
"mat"

The training process calculates an error.

The model then adjusts its internal parameters.

This happens repeatedly across huge amounts of training data.

## Important: What Happens During Training?

Training does NOT mean:

"Store every sentence exactly and retrieve it later."

Instead, the model adjusts numerical parameters to learn patterns from its training data.

The learned information is represented in the model's parameters.

This is one reason LLMs can generate new sentences instead of simply copying a training sentence every time.

## What Are Parameters?

Parameters are numerical values inside a neural network that are learned during training.

You can think of parameters as the model's learned internal settings.

During training:
```text
Data
 ↓
Model
 ↓
Prediction
 ↓
Error
 ↓
Parameter Updates
 ↓
Improved Model
```
Large language models can contain very large numbers of parameters.

### Simple analogy for parameters

Imagine learning to recognize a cat.

You see many examples:
```text
Cat
Cat
Cat
Dog
Cat
Dog
```
Your brain gradually learns useful patterns.

An artificial neural network does something conceptually similar through numerical parameters.

The model does not have a single parameter called:

"cat = animal"

Instead, information is distributed across many learned numerical values.

This analogy is not mathematically exact, but it is useful for understanding the basic idea.

## Parameters vs Hyperparameters

### Parameters

Learned by the model during training.

Examples:

- Neural network weights
- Neural network biases


### Hyperparameters

Chosen by developers or during the training setup.

Examples:

- Learning rate
- Batch size
- Number of training epochs
- Model architecture settings

Simple difference:

Parameters
→ Learned by the model

Hyperparameters
→ Set as part of the training process

## What is a Token?

A token is a unit of text processed by a language model.

A token may represent:

- A complete word
- Part of a word
- Punctuation
- A space-related piece

The exact tokenization depends on the tokenizer and model.

Example:

"machine learning"

might be represented using several tokens.

The important idea is:
```text
Human text
 ↓
Tokens
 ↓
Numbers
 ↓
LLM

## Tokenization Example

Sentence:

"I love machine learning."

The tokenizer might split it into pieces such as:

"I"
"love"
"machine"
"learning"
"."

But tokenization is model-dependent.

A tokenizer may also split words into smaller pieces.

For example:

"unbelievable"

could potentially be represented as multiple subword tokens.

Therefore:

1 word ≠ necessarily 1 token.

This distinction becomes very important when working with LLM APIs.

## Tokens → Token IDs

LLMs work with numerical representations.

Simplified example:

Text:

"I love AI"

↓

Tokens:

["I", "love", "AI"]

↓

Token IDs:

[125, 843, 921]

The actual numbers depend on the tokenizer.

The model processes numerical token representations rather than raw text characters.

## What is Context?

Context is the surrounding information available to the model when generating or interpreting a response.

Example:

User:

"Who is the CEO of Apple?"

The model needs the context of the question to produce an appropriate answer.

In a conversation:

User:
"My name is Rahul."

User:
"What is my name?"

The previous message provides context.

The model can use the conversation history that is included in its available context.

## What is a Context Window?

A context window is the amount of tokenized information a model can consider as input for a particular request.

It can include:

- System instructions
- User messages
- Previous conversation
- Documents
- Tool information
- Other supplied context

Simplified:
```text
Context
│
├── Instructions
├── Previous messages
├── Current question
└── Other information
        ↓
       LLM
        ↓
    Response
```
Different models support different context-window sizes.

## What is a Prompt?

A prompt is the input or instruction given to a language model.

Examples:

"Explain machine learning."

"Summarize this article."

"Write Python code to sort a list."

"Translate this sentence into French."

The prompt provides the task and, when needed, additional context or constraints.

**Simple definition**

**Prompt = The instruction/input we give to an LLM.**

## What is a Response?

A response is the output generated by the language model based on the provided input and context.

Example:

Prompt:

"Explain what NLP is."

Response:

"NLP is a field of AI that enables computers to work with human language."

So:
```text
Prompt
 ↓
LLM
 ↓
Response
```
The response is generated token by token based on the model's learned patterns and the provided context.

## Prompt → Response

Example:

Prompt:

"Explain neural networks in simple words."

        ↓

       LLM

        ↓

Response:

"A neural network is a machine learning model inspired
loosely by the way biological neurons process information."

The model uses:

- The prompt
- Available context
- Learned parameters

to generate the response.

## What Happens When We Send a Prompt?

Simplified process:
```text
User Prompt
     ↓
Tokenization
     ↓
Token IDs
     ↓
LLM
     ↓
Probability Distribution
     ↓
Next Token Selection
     ↓
More Tokens
     ↓
Complete Response
```
The model generates the response progressively rather than magically producing the entire paragraph in one thought.

## Next-Token Prediction

A fundamental idea behind language models is predicting likely next tokens.

Example:

"The capital of France is"

The model assigns probabilities to possible next tokens.

Simplified:
```text
Paris     → high probability
London    → low probability
Banana    → very low probability
```
The model selects a token according to its generation process.

Then it continues predicting the next token.

This process repeats until the response is complete.

## Important: Is an LLM Just Autocomplete?

At a basic level, language generation involves predicting the next token.

However, modern LLMs are much more capable than ordinary phone autocomplete.

They can perform tasks such as:

- Reasoning-like problem solving
- Summarization
- Translation
- Code generation
- Information extraction
- Following instructions
- Conversational interaction

These capabilities emerge from the scale, architecture, training data and training methods used to build modern models.

## Training vs Inference

These are two important terms.

### Training

The model learns its parameters from data.
```text
Data
 ↓
Training
 ↓
Learned Parameters
```

### Inference

A trained model is used to generate a prediction or response.
```text
Prompt
 ↓
Trained Model
 ↓
Response
```

#### Easy memory trick
Training → Learn

Inference → Use

## What is an LLM Hallucination?

An LLM can sometimes generate information that sounds convincing but is incorrect or unsupported.

This is commonly called a hallucination.

Example:

User:
"Give me the exact source for this claim."

The model may produce a plausible-looking but incorrect citation.

Therefore:

Confident language
≠
Guaranteed truth

LLM applications often need techniques such as:

- Retrieval-Augmented Generation (RAG)
- Tool use
- Grounding
- Validation
- Human review
