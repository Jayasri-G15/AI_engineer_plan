## 1. What Is a Chatbot?

A chatbot is a software application that communicates with a user through natural language.

A traditional chatbot may use predefined rules.

An AI chatbot uses an AI/LLM model to generate responses.

Example:

User:
"Explain machine learning"

Chatbot:
"Machine learning is a branch of AI..."

The chatbot receives the user's message, sends it to an LLM, receives the generated response, and displays it to the user.

### LLM chatbot
```text

````python
import time

def generate_response(prompt, max_retries=3):
    for attempt in range(max_retries):
        try:
            response = client.models.generate_content(
                model="gemini-3.6-flash",
                contents=prompt
            )

            return response.text

        except Exception as error:
            print(f"Attempt {attempt + 1} failed.")
            print(error)

            if attempt < max_retries - 1:
                wait_time = 2 ** attempt
                print(f"Retrying in {wait_time} seconds...")
                time.sleep(wait_time)
            else:
                print("All attempts failed.")
                return None
````

````python
answer = generate_response(
    "how can i reduce my skin tan instantly?"
)

print("Bot:", answer)
````

## 19. Why Is This Not Yet a Good Chatbot?

Our chatbot currently forgets previous messages.

Example:

User:
"My name is Rahul."

Bot:
"Nice to meet you!"

User:
"What is my name?"

The model may not know the answer because we did not send the previous conversation.

We need:

Conversation History

## 20. Conversation History

A chatbot can maintain previous messages in Python.

Example:
```python
history = [
    {
        "role": "user",
        "content": "My name is Rahul."
    },
    {
        "role": "assistant",
        "content": "Nice to meet you!"
    }
]
```
When the next request is sent, relevant conversation history can be included.

This gives the model context.

## 21. Context

The model can answer correctly if the previous conversation is available as context.

## 24. Chatbot Loop

A real chatbot usually runs continuously.

Flow:
```python
while True:

    receive user input

    if user wants to exit:
        stop
```
```text
    send message to LLM

    receive response

    display response

This creates an interactive conversation.

````python
while True:
    user_message = input("You: ")

    if user_message.lower() in ["exit", "quit", "bye"]:
        print("Bot: Goodbye!")
        break

    response = client.models.generate_content(
        model="gemini-3.7-flash",
        contents=user_message
    )

    print("Bot:", response.text)
````

## 25. Why Do We Need Error Handling?

API requests can fail.

Possible reasons:

- Invalid API key
- Network problem
- Rate limit
- Invalid request
- Service unavailable
- Model unavailable

A production chatbot should not crash whenever an API hiccups.

````python
try:
    user_message = input("You: ")

    response = client.models.generate_content(
        model="gemini-3.7-flash",
        contents=user_message
    )

    print("Bot:", response.text)

except Exception as error:
    print("Something went wrong.")
    print("Error:", error)
````

## 29. Gemini vs Ollama vs NVIDIA

### Gemini

Cloud API.

Advantages:

- Easy setup
- Free tier
- No local GPU required
- Good for learning API integration

### Ollama

Runs models locally.

Advantages:

- Local
- Private
- No API cost for local inference
- Good for experimentation

Disadvantage:

Performance depends on your computer.

### NVIDIA NIM

Provides optimized model inference and APIs.

Advantages:

- Strong NVIDIA ecosystem
- GPU optimized
- Useful for advanced AI applications
- API-based and self-hosting options

For our first chatbot:

Gemini → easiest starting point

Later:

Ollama → local chatbot

NVIDIA → advanced API/deployment exploration
