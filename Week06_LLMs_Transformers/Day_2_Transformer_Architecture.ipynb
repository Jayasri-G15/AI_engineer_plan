## What is a Transformer?

A Transformer is a deep learning architecture designed to process sequences of data, especially language.

Transformers became extremely important in NLP because they can process relationships between words using an attention mechanism.

The Transformer architecture was introduced in the 2017 research paper:

"Attention Is All You Need"

Transformers became the foundation of many modern language models and other AI systems.

#### Simple definition

**A Transformer is a neural-network architecture that uses attention mechanisms to understand relationships between elements in a sequence.**

## Why Were Transformers Needed?

Before Transformers, NLP systems commonly used architectures such as:

- RNNs
- LSTMs
- GRUs

These models processed sequences step by step.

Example:

"The cat sat on the mat."

The model would process the sequence progressively.

This made it harder to efficiently handle long-range relationships and large-scale training.

Transformers introduced a different approach using attention.

Instead of relying only on sequential processing, attention allows the model to consider relationships between different parts of the input.

## Before Transformers — RNN Idea

Consider:

"The animal didn't cross the road because it was tired."

To understand "it", the model needs information from another part of the sentence.

A sequential model processes words one after another.

For long sentences, remembering relationships across distant words can become difficult.

Transformers address this using attention mechanisms.

## Basic Transformer Pipeline
```text
Input Text
    ↓
Tokenization
    ↓
Tokens
    ↓
Embeddings
    ↓
Transformer
    ↓
Output
```
Example:
```text
"I love machine learning"

        ↓

["I", "love", "machine", "learning"]

        ↓

Numerical vectors

        ↓

Transformer layers

        ↓

Output representation / predicted tokens

## Step 1 — Input

The process begins with text.

Example:

"I love machine learning."

This is the human-readable input.

The Transformer cannot directly operate on the raw sentence.

The text must first be converted into a representation that the neural network can process.

## Step 2 — Tokenization

The input text is divided into tokens.

Example:

"I love machine learning."

could become:

["I", "love", "machine", "learning", "."]

The exact tokens depend on the tokenizer.

A word may sometimes be represented by multiple tokens.

After tokenization, tokens are converted into token IDs.

Example:

["I", "love", "machine"]

        ↓

[101, 245, 893]

The actual IDs depend on the tokenizer.

## Step 3 — Embeddings

Token IDs are not meaningful numerical values by themselves.

The model converts tokens into dense numerical vectors called embeddings.

Example:

"cat"

↓

[0.21, -0.45, 0.72, 0.13, ...]

Each token receives a vector representation.

These vectors allow the neural network to process information about the tokens.

Pipeline:

Token
 ↓
Token ID
 ↓
Embedding Vector

## What is an Embedding?

An embedding is a numerical vector representation of an item such as a word or token.

The vector contains learned numerical information that allows the model to represent relationships between tokens.

For example:

"king"

"queen"

"man"

"woman"

can have relationships in the learned representation space.

Embeddings are not simply dictionary definitions.

They are learned numerical representations.

## Token ID vs Embedding

Token ID:
```text
"cat"
 ↓
1257
```
This is simply an identifier.

Embedding:
```text
"cat"
 ↓
[0.21, -0.14, 0.73, 0.08, ...]
```
This is a vector representation used by the neural network.

Therefore:

Token ID
→ Identifier

Embedding
→ Numerical representation used by the model

## Positional Information

Language depends on word order.

Compare:

"Dog bites man."

"Man bites dog."

The same words appear, but the meaning is different.

Therefore, a Transformer needs information about the positions of tokens.

A Transformer uses positional information so that the model can distinguish:
```text
Token 1
Token 2
Token 3
...
```
from the same tokens appearing in different positions.

## Step 4 — Transformer

The embeddings are passed into Transformer layers.

The Transformer processes the token representations using mechanisms such as:

- Attention
- Feed-forward neural networks
- Normalization
- Residual connections

The most important concept for today is:

# Attention

Attention allows tokens to consider information from other tokens.

## Simple Attention Example

Consider:

"The animal crossed the road because it was tired."

What does "it" refer to?

A useful language model needs to understand the relationship between:

"it"

and

"animal"

Attention helps the model determine which other tokens are relevant to the current token.

This is the basic intuition behind attention.

## What is Self-Attention?

Self-attention allows each token to consider other tokens within the sequence.

Example:

"The cat sat on the mat."

When processing:

"cat"

the model can consider other tokens in the sentence.

When processing:

"mat"

the model can also consider other relevant tokens.

This allows the model to build contextual representations of tokens.

### Simple idea:
```text
Token
 ↓
Look at other relevant tokens
 ↓
Combine useful information
 ↓
Updated representation

## Why Does Attention Matter?

Consider:

"The bank approved the loan."

The word "bank" could refer to different concepts.

Context helps determine its meaning.

Compare:

"The bank approved the loan."

"The boat reached the river bank."

The surrounding words provide important information.

Attention allows the model to consider surrounding tokens when creating contextual representations.

## What is an Encoder?

An Encoder takes an input sequence and creates contextual representations of that input.

Simplified:
```text
Input
 ↓
Tokenization
 ↓
Embeddings
 ↓
Encoder
 ↓
Contextual Representations
```
The encoder focuses primarily on understanding or representing the input.

### Examples of encoder-style models
BERT

RoBERTa

These are commonly associated with encoder-based Transformer architectures.

## Encoder Example

Input:
```text
"The movie was excellent."

        ↓

Tokenizer

        ↓

Tokens

        ↓

Embeddings

        ↓

Encoder

        ↓

Contextual representation

        ↓

Classification / extraction / other task
```

An encoder is particularly useful when we need to understand the input.

## What is a Decoder?

A Decoder generates output based on the information available to it.

In a language-generation setting, it predicts the next token.

Example:

Input:
```text
"The capital of France is"

        ↓

Decoder

        ↓

"Paris"
```
Then the model can continue generating additional tokens.

### Simple idea:
```text
Encoder
→ Understand / represent input

Decoder
→ Generate output
```
This is simplified, but useful for learning the architecture.

## Encoder-Decoder Transformer

The original Transformer architecture contains:

Encoder
+
Decoder

Simplified:
```text
Input
 ↓
Encoder
 ↓
Representation
 ↓
Decoder
 ↓
Output
```
This architecture is useful for sequence-to-sequence tasks.

Examples:
```text
-Translation
-Summarization
-Text transformation

## Encoder-Decoder Example — Translation

Input:
```text
"Bonjour tout le monde"

        ↓

Encoder

        ↓

Contextual representation

        ↓

Decoder

        ↓

"Hello everyone"
```
The encoder processes the source sentence.

The decoder generates the target sentence.

## Encoder vs Decoder

| Feature | Encoder | Decoder |
|---|---|---|
| Main role | Understand/represent input | Generate output |
| Input | Input sequence | Previously generated tokens + context |
| Common use | Classification, representation | Text generation |
| Example | BERT | GPT-style models |

Simple memory:

Encoder
→ Understand

Decoder
→ Generate

## Autoregressive Generation

Decoder-based language models commonly generate text one token at a time.

Example:

Prompt:

"The cat"

Step 1:

"The cat"

→ predicts "is"

Step 2:

"The cat is"

→ predicts "sleeping"

Step 3:

"The cat is sleeping"

→ predicts the next token

This continues until the model reaches an appropriate stopping condition.

This connects directly to yesterday's **next-token prediction** concept.

## Attention Inside a Transformer

A simplified Transformer layer can be viewed as:
```text
Input Embeddings
       ↓
Self-Attention
       ↓
Add & Normalize
       ↓
Feed-Forward Network
       ↓
Add & Normalize
       ↓
Output
```
A Transformer contains multiple such layers.

Each layer progressively transforms the token representations.

## Multi-Head Attention

Transformers commonly use multiple attention heads.

Instead of having one attention mechanism:

        Attention

they use multiple:
```text
Attention Head 1
Attention Head 2
Attention Head 3
...
Attention Head N
```
Each head can learn different relationships between tokens.

The results are combined to create a richer representation.

## Query, Key and Value — Preview

Attention is commonly implemented using three concepts:
```text
Query
Key
Value
```
A simplified idea:

Query
→ What information am I looking for?

Key
→ What information does each token represent?

Value
→ What information should be passed forward?

Attention compares queries and keys to determine how strongly tokens should interact, then uses the values to build the resulting representation.


# Complete Transformer Flow
```text
Input Text
     ↓
Tokenization
     ↓
Token IDs
     ↓
Embeddings
     ↓
Positional Information
     ↓
Transformer Layers
     │
     ├── Self-Attention
     │
     ├── Feed-Forward Network
     │
     ├── Normalization
     │
     └── Residual Connections
     │
     ↓
Output Representation
     ↓
Prediction / Generated Output
```
This is the core architecture you should understand

## How Does This Connect to LLMs?

Modern LLMs are commonly built using Transformer architectures.

Simplified:
```text
User Prompt
     ↓
Tokenizer
     ↓
Token IDs
     ↓
Embeddings
     ↓
Transformer Layers
     ↓
Output Probabilities
     ↓
Next Token
     ↓
Next Token
     ↓
Next Token
     ↓
Generated Response
```
This is why understanding Transformers is essential for understanding LLMs.

## Why Did Transformers Become So Important?

Transformers brought several important advantages.

### 1. Attention

They can model relationships between different tokens.

### 2. Parallel processing during training

Much of the sequence can be processed in parallel during training, which makes large-scale training more practical.

### 3. Scalability

Transformer architectures can be scaled to very large models and datasets.

### 4. Strong language representations

They can learn rich contextual relationships between tokens.

These properties helped make modern LLMs possible.

## Traditional NLP vs Transformer

Traditional approach:
```text
Text
 ↓
Preprocessing
 ↓
TF-IDF
 ↓
ML Model
 ↓
Prediction
```

Transformer approach:
```text

Text
 ↓
Tokenization
 ↓
Embeddings
 ↓
Attention
 ↓
Transformer Layers
 ↓
Output
```
Traditional NLP still has useful applications.

Transformers simply provide a much more powerful architecture for many language tasks.
