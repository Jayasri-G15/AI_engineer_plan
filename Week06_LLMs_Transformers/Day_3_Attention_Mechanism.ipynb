### Simple definition

**Attention allows a model to focus more on relevant parts of the input.**

## Human Analogy

Imagine reading:

"John went to the store because he needed milk."

When you see:

"he"

you naturally connect it with:

"John"

You don't treat every word as equally important.

You focus on the words that help determine the meaning.

Attention gives a neural network a mathematical mechanism for learning similar relationships.


This is an analogy, not a claim that the model literally reads like a human.

## Simple Attention Example

Sentence:

"The cat sat on the mat."

Suppose we are processing:

"cat"

Potentially relevant words include:
```text
"the"
"sat"
"mat"
```
The model calculates attention relationships between tokens.

A simplified representation might look like:
```text
cat → the   : 0.10
cat → sat   : 0.45
cat → mat   : 0.30
```
Higher value
→ More attention

Lower value
→ Less attention


The actual calculations in a Transformer are much more sophisticated.

## Attention Weights

Attention produces weights representing how strongly one token should consider other tokens.

Example:

"The animal was tired because it ran all day."

For the token:

"it"

the model may assign different weights to surrounding tokens.

Conceptually:
```text
animal → high relevance
tired  → some relevance
ran    → some relevance
day    → some relevance
```
The model uses these relationships to create a better contextual representation.

#### Important:

The exact attention weights are learned by the model and depend on the input and model architecture.

## What is Self-Attention?

Self-attention is a type of attention where tokens in the same sequence attend to one another.

Example:

"The cat chased the mouse."

When processing "cat", the model can consider other tokens.

When processing "mouse", the model can also consider other tokens.

Therefore:
```text
Token
 ↓
Other tokens in the same sequence
 ↓
Relevant information
 ↓
Updated representation
```
### Simple definition

**Self-attention allows tokens to consider other tokens within the same sequence.**

## Why Is It Called "Self"-Attention?

It is called self-attention because the attention operation is performed within the same sequence.

For example:

Input:

"The cat sat on the mat."

The tokens attend to other tokens from that same input sequence.

So:
```text
Sequence
 ↓
Tokens attend to tokens
 ↓
Self-Attention

## Self-Attention Example

Sentence:

"The dog chased the ball because it was moving."

Suppose we are processing:

"it"

The model can compare "it" with other tokens:
```text
dog
chased
ball
moving
```
The relationships learned through attention help build a contextual representation of "it".

The model is not simply looking at the word itself.

It considers the surrounding sequence.

## What is Context?

Context is the surrounding information that helps determine the meaning of a piece of text.

Example:

"I went to the bank."

By itself, "bank" is ambiguous.

Compare:

"I went to the bank to deposit money."

versus:

"I sat on the bank of the river."

The surrounding words provide context that helps determine the meaning.

## Context in a Conversation

Consider:

User:
"My name is Alex."

Assistant:
"Nice to meet you, Alex."

User:
"What is my name?"

The previous message provides context.

Without the relevant previous information:

"What is my name?"

is much harder to answer correctly.

Therefore, conversational AI systems need mechanisms for managing conversation context.

## Relationship Between Attention and Context

Attention helps the model build contextual representations.

Simplified:
```text
Tokens
 ↓
Self-Attention
 ↓
Relationships between tokens
 ↓
Contextual representations
 ↓
Understanding / Generation
```
So:

**Context** is the information surrounding the current input.

**Attention** is one of the mechanisms used by Transformers to determine relationships within that information.

## What is a Context Window?

A context window is the maximum amount of tokenized information that a model can consider as input for a particular request.

The context can include:

- System instructions
- Previous conversation
- Current user message
- Documents
- Tool outputs
- Other supplied information

Simplified:
```text

┌───────────────────────────────┐
│         CONTEXT WINDOW        │
│                               │
│ Instructions                  │
│ Previous messages             │
│ Current prompt                │
│ Documents                     │
│ Other context                 │
│                               │
└───────────────────────────────┘
                ↓
               LLM
                ↓
             Response

## Context Window Is Not the Same as Memory

These concepts are often confused.

### Context Window

Information currently supplied to the model for a request.

### Long-Term Memory

Information stored separately and retrieved when needed.

For example:

A chatbot might store:

"User prefers Python."

in a database.

When the user asks a future question, the application can retrieve that information and place it into the model's context.

Therefore:
```text
Stored information
        ↓
Application retrieves it
        ↓
Added to context
        ↓
LLM processes it

## Context Window Is Not the Same as Memory

These concepts are often confused.

### Context Window

Information currently supplied to the model for a request.

### Long-Term Memory

Information stored separately and retrieved when needed.

For example:

A chatbot might store:

"User prefers Python."

in a database.

When the user asks a future question, the application can retrieve that information and place it into the model's context.

Therefore:
```text
Stored information
        ↓
Application retrieves it
        ↓
Added to context
        ↓
LLM processes it

The exact token limit depends on the specific model.

## Why Do Token Limits Matter?

Imagine a chatbot conversation:
```text
Message 1
Message 2
Message 3
Message 4
...
Message 1000
```
The conversation keeps growing.

If all previous messages are included in every request, the context becomes larger and larger.

Eventually, the application may need to remove, summarize, or selectively retrieve information.

Therefore:
```text
Long conversation
        ↓
More tokens
        ↓
Larger context
        ↓
Context management becomes important

## Example

Suppose a hypothetical model has:

10,000 token context window.

A request contains:

System instructions → 1,000 tokens

Conversation → 5,000 tokens

Document → 3,000 tokens

User question → 500 tokens

Total:

9,500 tokens

This fits within the hypothetical limit.

If another 1,000 tokens are added:

Total:

10,500 tokens

Now the input exceeds the hypothetical context window.

The application needs to reduce or manage the context.

## Context Window and Output

In many LLM systems, the overall context budget relates to both input and generated output.

Simplified:

Input Tokens
+
Output Tokens
=
Total Token Usage

Example:

Input:
7,000 tokens

Requested/generated output:
2,000 tokens

Total:
9,000 tokens

Whether this fits depends on the model's limits and API rules.

Always check the documentation for the specific model being used.

## What Happens With Large Documents?

Suppose we have:

A 500-page PDF.

We cannot simply assume the model should receive the entire document every time.

A common strategy is:
```text
Large Document
      ↓
Split into chunks
      ↓
Create representations
      ↓
Find relevant chunks
      ↓
Add relevant chunks to context
      ↓
LLM
      ↓
Answer
```
This idea becomes the foundation for RAG

## Attention vs Context

These two terms are related but different.

### Attention

A mechanism inside the Transformer.

It determines how strongly tokens should interact with other tokens.

### Context

The information available to the model for the current request.

Simple distinction:

Attention
→ How the model focuses on relationships.

Context
→ What information the model has available.


## Memory trick 
Context = Information available

Attention = How relevant information is connected

## Context Window vs Token Limit

### Context Window

The model's available capacity for tokenized context.

### Token Limit

The maximum token-related amount allowed by a particular model/API constraint.

In practical conversations, people often use these terms loosely.

The exact limits depend on the model and service being used.

Always check the specific model documentation when building an application.

## How Does a Model Determine Which Words Are Important?

At a high level:

1. Tokens are converted into numerical representations.

2. The model creates Query, Key and Value representations.

3. Queries and Keys are compared.

4. The resulting scores determine attention weights.

5. Values are combined using those weights.

6. The result becomes a contextual representation.

Simplified:
```text
Tokens
 ↓
Queries / Keys / Values
 ↓
Attention Scores
 ↓
Attention Weights
 ↓
Weighted Values
 ↓
Contextual Representation

## Query, Key and Value — Simple Intuition

Think of a search system.

### Query

"What information am I looking for?"

### Key

"What information does each token represent?"

### Value

"What information should be passed forward?"

Attention compares the Query with Keys to determine relevance.

Then it uses the Values to construct the resulting representation.

## Attention Flow

Suppose the input is:

"The cat chased the mouse."

The model processes token representations.

For a token such as:

"cat"

it can calculate relationships with:
```text
"The"
"chased"
"the"
"mouse"
```
Conceptually:
```text
          The
           ↘
cat  ←─── chased
           ↘
          mouse
```
The attention mechanism learns which relationships are useful.

## Multi-Head Attention

Transformers generally use multiple attention heads.

Each attention head can learn different relationships.

For example, one head might learn relationships related to:

- Syntax
- Word relationships
- Long-distance dependencies
- Other patterns

Simplified:
```text
Input
 ↓
┌───────────────┐
│ Attention 1   │
│ Attention 2   │
│ Attention 3   │
│ ...           │
│ Attention N   │
└───────────────┘
 ↓
Combined Representation

## Why Multiple Attention Heads?

Language contains many kinds of relationships.

Example:

"The boy who won the race was extremely happy."

Different relationships may matter:

"boy" ↔ "was"

"won" ↔ "race"

"boy" ↔ "happy"

Multiple attention heads allow the model to learn different patterns simultaneously.

The outputs of the heads are combined.

## Causal Attention

Decoder-based language models commonly use causal attention.

The model can use previous tokens when predicting the next token.

Example:

"The cat"

The model can use:

"The"

to help predict:

"cat"

But while generating the next token, it should not look ahead at tokens that have not yet been generated.

Conceptually:

The → cat → sat → on → the → mat

When predicting:

"sat"

the model can attend to earlier tokens.

This supports autoregressive generation.

## Self-Attention vs Causal Attention

### Self-Attention

Tokens can attend to other tokens in the sequence, depending on the architecture and masking.

### Causal Self-Attention

A decoder prevents a token from attending to future tokens.

Example:

When predicting token 3:
```text
Token 1 ✓
Token 2 ✓
Token 3 → current position
Token 4 ✗
Token 5 ✗
```
This prevents the model from seeing future information during autoregressive generation.

## Context Management in AI Applications

When building an LLM application, we may need to manage context.

Common techniques:

### 1. Truncation

Remove older or less important messages.

### 2. Summarization

Summarize previous conversation.

### 3. Chunking

Split large documents into smaller pieces.

### 4. Retrieval

Retrieve only relevant information.

### 5. Memory systems

Store useful information externally and retrieve it when needed.

## Why Context Is Critical

An LLM can have strong language capabilities but still produce a poor answer if the relevant information is missing.

Example:

Question:

"What is the status of order #1234?"

Without order data:

The model does not automatically know the real-time order status.

With retrieved context:

Order #1234
Status: Shipped

The model can use that information to generate a useful answer.

Therefore:

LLM

Relevant Context

More useful AI application
