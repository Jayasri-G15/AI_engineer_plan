## What is an API?

API stands for:

Application Programming Interface

An API allows one software application to communicate with another software system.

For example:
```text
Python Application
       ↓
      API
       ↓
LLM Service
```
### Simple definition

**An API is a way for software systems to communicate with each other.**

## Simple API Analogy

Imagine a restaurant.

You:

→ Customer

Kitchen:

→ Service/System

Waiter:

→ API

You tell the waiter what you want.

The waiter sends the request to the kitchen.

The kitchen prepares the food.

The waiter brings the result back.

Software works similarly:
```text
Your Python program
        ↓
API Request
        ↓
LLM Service
        ↓
API Response
        ↓
Your Python program
```
The API is the communication interface between your application and the service

## Why Do We Need an API?

Suppose you want to use an LLM in your Python application.

Your application needs to send:

- A prompt
- Model information
- Other request parameters

to the LLM service.

The API provides a standard way to send this information and receive the generated result.

Without an API, your application would not have the standard interface needed to communicate with the remote service.

## LLM API Architecture

Your Application
```text
Python
  │
  │ API Request
  ▼
┌───────────────────┐
│    LLM API        │
│                   │
│ Authentication    │
│ Request handling  │
│ Model execution   │
└───────────────────┘
  │
  │ API Response
  ▼
Python Application
```
Your Python application doesn't need to run the entire LLM itself.

It can communicate with a service that hosts the model.

## What is an API Request?

A request is the information your application sends to an API.

An LLM request can contain things such as:

- API authentication
- Model name
- User prompt
- Conversation messages
- Parameters
- Other configuration

Simplified:
```text
Python
 ↓
Request
{
    prompt: "Explain NLP"
}
 ↓
LLM API

## What is an API Response?

A response is the information returned by the API after processing the request.
```text
Simplified:

Request:

"Explain NLP."

        ↓

      LLM API

        ↓

Response:

"NLP is a field of AI that allows computers
to process human language."
```
Python receives this response and can use it in the application.

## Request vs Response

### Request

Application → API

Contains information asking the service to perform something.

### Response

API → Application

Contains the result returned by the service.

Memory trick:

Request
→ "Please do this."

Response
→ "Here is the result."

## What is an API Key?

An API key is a credential used by an API service to identify and authorize an application or user.

Simplified:
```text
Python Application
       ↓
API Key
       ↓
LLM API
       ↓
Authentication
       ↓
Request processed
```
An API key is similar to a password for accessing a service, although the exact authentication mechanism depends on the API.

## Why Are API Keys Important?

An API provider may use API keys for:

- Authentication
- Authorization
- Usage tracking
- Rate limiting
- Billing/account association

For example:

Application A
→ API Key A

Application B
→ API Key B

The service can distinguish requests from different credentials.

## API Key Security

Never hard-code a real API key directly into application code.

Bad:

api_key = "my-real-secret-key"

This can accidentally expose the key through:

- GitHub
- Screenshots
- Shared notebooks
- Logs
- Public repositories
- Chat messages

Instead, store secrets outside your source code.

## What is an Environment Variable?

An environment variable is a value stored outside your application code and provided through the operating environment.

Common examples:

```text
API_KEY
DATABASE_URL
SECRET_KEY
PORT
ENVIRONMENT
```
For an API key:
```text
Environment
     ↓
API_KEY
     ↓
Python Application
```
This helps separate configuration/secrets from source code.

## Using a .env File

During local development, a common approach is to store environment variables in a `.env` file.

Example:

API_KEY=your_api_key_here

The application loads the value from the environment.

Important:

The `.env` file should normally NOT be committed to Git.

## Protecting .env with .gitignore

If your project uses Git, add `.env` to `.gitignore`.

Example:

.env

This helps prevent accidentally committing secrets to the repository.

A common project structure:
```text
project/
│
├── app.py
├── .env
├── .gitignore
└── requirements.txt

## What is JSON?

JSON stands for:

JavaScript Object Notation

It is a common format for exchanging structured data between applications.

Example:

{
    "name": "Alex",
    "age": 25
}

APIs commonly use JSON for requests and responses.

A simplified LLM request could look conceptually like:
```python
{
    "model": "some-model",
    "messages": [
        {
            "role": "user",
            "content": "Explain NLP"
        }
    ]

The API receives structured information describing what we want it to do.

The exact fields depend on the API provider and API version.

## Example API Response

An API may return structured JSON similar to:
```python
{
    "output": "NLP is a field of AI..."

Your Python application can read the returned data.

     ↓

     ↓

The exact response structure depends on the API being used.

## HTTP Status Codes

APIs commonly use HTTP status codes to indicate what happened.

Examples:

200
→ Request succeeded

400
→ Bad request

401
→ Authentication problem

403
→ Access forbidden

404
→ Resource not found

429
→ Rate limit exceeded

500
→ Server error

These codes are extremely useful when debugging API integrations.

## Request Headers

HTTP requests can contain headers.

Headers provide additional information about the request.

For authenticated APIs, a request may contain an authorization header.

Conceptually:

Headers:

Authorization: API credential
Content-Type: application/json

Body:
```python
{
    "prompt": "Explain NLP"
}
```
The exact authentication format depends on the API provider.

## Calling an API from Python

Python can make HTTP requests using libraries such as:

- requests
- httpx

For example:

```python
import requests

response = requests.get("https://example.com")

print(response.status_code)
```

This demonstrates the basic idea of:
```text
Python
↓
HTTP Request
↓
Server
↓
HTTP Response
```

This is only a generic API example, not an LLM API call.

````python
import requests

url = "https://example.com/api"

payload = {
    "message": "Hello"
}

response = requests.post(
    url,
    json=payload
)

print("Status code:", response.status_code)
print("Response:", response.text)
````

The URL above is only an example.

For a real LLM integration, we will use the API endpoint and SDK provided by the model provider.

## What is an SDK?

SDK stands for:

Software Development Kit

An SDK provides programming tools that make it easier to interact with a service.

Instead of manually constructing every HTTP request, an SDK may provide Python methods/classes for the API.

Conceptually:

Without SDK:
```text
Python
 ↓
HTTP
 ↓
JSON
 ↓
API
```

With SDK:
```text
Python
 ↓
SDK
 ↓
API
```
Both approaches ultimately communicate with the service's API.

## SDK vs Direct HTTP

### Direct HTTP

You manage:

- URLs
- Headers
- Authentication
- JSON
- HTTP requests
- Error handling

### SDK

The SDK handles much of the API communication for you.

You typically write simpler Python code.

For application development, SDKs are often more convenient.

Understanding HTTP is still important because it helps you debug what is happening underneath.

## API Error Handling

API calls can fail.

Possible reasons:

- Invalid API key
- Missing API key
- Incorrect request
- Invalid model
- Rate limit
- Network problem
- Server problem
- Insufficient account access

A production application should handle these errors gracefully.

Instead of:

"Everything exploded."

we want:

"LLM service is temporarily unavailable. Please try again."

A surprisingly important distinction in software engineering.

````python
import requests

try:
    response = requests.get("https://example.com")
    response.raise_for_status()

    print("Request successful")
    print(response.text)

except requests.RequestException as error:
    print("API request failed:", error)
````

This is a generic example of handling HTTP errors.
