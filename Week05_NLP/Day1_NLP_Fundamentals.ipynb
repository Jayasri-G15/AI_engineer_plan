## 1. What is NLP?

NLP stands for **Natural Language Processing**.

NLP is a branch of Artificial Intelligence that enables computers to work with human language.

Human beings communicate using languages such as:

- English
- Tamil
- Hindi
- Spanish
- French
- and many others.

Computers, however, work with numerical and structured representations of information.

NLP provides techniques that allow computers to process, analyze, understand, and generate human language.

### Simple Definition

> NLP is the field of AI that helps computers understand and work with human language.

### Example

Human:

"I really enjoyed this movie."

An NLP system can process this sentence and determine that the person expressed a **positive opinion**.

### Basic Flow
```text
Human Language
↓
NLP
↓
Computer Processing
↓
Understanding / Prediction / Response

## Simple NLP Example

Consider the sentence:

> "I love this phone."

An NLP system can identify:

- "I" → Person
- "love" → Action / sentiment-related word
- "phone" → Object

It can also determine that the sentence expresses a **positive opinion**.

Another example:

> "The customer is unhappy with the service."

An NLP system can identify that the sentence expresses a **negative sentiment**.

Therefore, NLP allows computers to extract useful information from human language.

## 4. What is Text Data?

Text data is information represented using words, sentences, or other written language.

Examples:

```text
"Hello, how are you?"

"I bought this product yesterday."

"The movie was amazing."

"Where is my order?"
```
Text data is usually called unstructured data because it does not naturally fit into fixed rows and columns like a traditional numerical dataset.

### Examples of Text Data:
```text
Customer reviews
Emails
Tweets/social media posts
Articles
Books
Chat messages
Product reviews
Support tickets
Comments
```
NLP techniques are used to convert this raw text into information that machine learning models can work with

## Structured vs Unstructured Data

### Structured Data

Data organized into rows and columns.

Example:

| Name | Age | Salary |
|---|---:|---:|
| John | 25 | 30000 |
| Priya | 28 | 40000 |

A computer can easily perform calculations on these values.

### Unstructured Data

Information such as:

```text
"I purchased this phone last week and I absolutely love it."
```
The computer cannot directly treat the sentence like numerical columns.

NLP helps convert the language into useful representations.

## 5. Natural Language Understanding (NLU)

NLU stands for **Natural Language Understanding**.

It is a part of NLP that focuses on helping computers understand the meaning and intent behind human language.

### Example

Consider:

> "Can you tell me today's weather?"

A computer should understand that the user is asking for:

- Information about weather
- For today

The important concept is **intent**.

### Another Example

User:

> "I want to cancel my order."

The system should understand:

Intent → Cancel Order

It should not simply treat the sentence as a collection of individual words.

### NLU can involve understanding:

- Meaning
- Intent
- Entities
- Context
- Relationships between words

## Intent

Intent means the purpose or goal behind what a user says.

Example:

User:

> "Where is my order?"

Intent:

> Track Order

User:

> "I want to return this product."

Intent:

> Product Return

User:

> "I forgot my password."

Intent:

> Password Reset

The system uses language understanding techniques to determine the user's intent.

## Entities

An entity is an important piece of information contained in a sentence.

Example:

> "Book a flight to Chennai tomorrow."

Possible entities:

- Chennai → Location
- Tomorrow → Date
- Flight → Travel-related entity

Another example:

> "I want to order an iPhone 17."

Entity:

- iPhone 17 → Product

Extracting important entities helps an NLP system understand what the user is talking about.

## Context in NLP

Understanding a sentence sometimes requires information from previous sentences.

Example:

User:

> "I bought a new phone yesterday."

Then:

> "It is not working."

The word "It" refers to the phone.

A system needs context to understand what "It" refers to.

Therefore, understanding language is not always just about individual words. Context can also be important.

# 6. How Does a Computer Process a Sentence?

A computer cannot understand human language in exactly the same way a human does.

The text must be converted into representations that a computer can process.

A simplified NLP pipeline is:
```text
Human Sentence
↓
Text Cleaning / Preprocessing
↓
Tokenization
↓
Numerical Representation
↓
Feature / Semantic Representation
↓
NLP Model
↓
Prediction / Understanding

# Example Pipeline

Sentence:

> "I love this movie!"

### Step 1 — Input

The computer receives:

"I love this movie!"

### Step 2 — Preprocessing

The text may be cleaned or normalized.

Example:

"I love this movie!"

→

"i love this movie"

### Step 3 — Tokenization

The sentence can be divided into tokens:

```text
["i", "love", "this", "movie"]
```
### Step 4 — Numerical Representation

Machine learning models generally work with numerical representations rather than raw words.

The words can therefore be represented using numerical techniques such as vectors/embeddings.

### Step 5 — Model Processing

The model processes the representation and learns patterns.

### Step 6 — Output

For a sentiment-analysis task, the output could be:
```text
Positive
```
This is a simplified explanation. The exact representation and model architecture depend on the NLP technique being used.

## 7. Important NLP Terms

### Corpus

A collection of text documents used for analysis or training.

Example:

10,000 customer reviews = a corpus.

### Document

A single piece of text.

Example:

One customer review.

### Sentence

A group of words expressing an idea.

Example:

"I love this product."

### Token

A unit produced when text is divided into smaller pieces.

Example:

"I love this product"

→

"I", "love", "this", "product"

### Vocabulary

The collection of unique tokens/words used in a dataset.

### Context

Information surrounding a word or sentence that helps determine its meaning.

### Intent

The purpose behind a user's statement.

### Entity

An important piece of information mentioned in text.

## NLP vs Traditional Numerical Data

Traditional machine learning data might look like:

```text
Age     Salary    Experience

30      50000     5
```
These values are already numerical.

#### Text data looks like:
```text
"I really enjoyed the product."
```
The computer cannot directly perform the same numerical calculations on the raw sentence.

Therefore, NLP uses techniques to transform language into representations that machine learning algorithms can process.

This is why preprocessing, tokenization and numerical representation are important in NLP.
