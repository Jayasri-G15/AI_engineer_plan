## 1. What is Text Classification?

Text classification is the process of assigning a piece of text to a predefined category or class.

In simple words:

Text → Machine Learning Model → Category

Example:
```text

        ↓
    Positive

"I hated this movie"
        ↓
    Negative
```
The machine learning model learns patterns from previously labelled text and then uses those patterns to classify new text.

## Step 3 — Feature Extraction

## What is a Feature?

A feature is a numerical representation of useful information from the text.

For example, suppose our vocabulary is:

["good", "bad", "love", "hate", "movie"]

For the sentence:

"I love this movie"

we could represent it using numbers.

The exact representation depends on the feature extraction technique.

Common techniques include:

1. Bag of Words
2. TF-IDF
3. Word Embeddings
4. Transformer embeddings

For today's traditional ML example, we will mainly use Bag of Words and TF-IDF.

## Bag of Words

Bag of Words is a simple technique for converting text into numbers.

Suppose we have:

Sentence 1:
"I love this movie"

Sentence 2:
"I hate this movie"

Vocabulary:

["I", "love", "hate", "this", "movie"]

Each sentence can be represented by counting how many times each word appears.

The model does not understand the meaning of the sentence.

It mainly looks at word occurrence.

````python
#Bag of Words with CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer

texts = [
    "I love this movie",
    "I hate this movie"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(texts)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nNumerical representation:")
print(X.toarray())
````

## Understanding the Bag of Words Output

Our vocabulary is:

["hate", "love", "movie", "this"]

Sentence:

"I love this movie"

becomes:

[0, 1, 1, 1]

Why?
```text
hate  → 0
love  → 1
movie → 1
this  → 1
```

Sentence:

"I hate this movie"

becomes:

[1, 0, 1, 1]

Why?
```text
hate  → 1
love  → 0
movie → 1
this  → 1
```

This is the basic idea behind feature extraction.

## TF-IDF

#### TF-IDF stands for:

Term Frequency — Inverse Document Frequency

It is another method for converting text into numerical features.

#### The basic idea is:

Words that are important to a particular document should receive higher importance.

Words that appear in almost every document may receive lower importance.

#### For example:

"movie"

may appear in many movie reviews.

A more specific word may be more useful for classification.

TF-IDF gives different numerical weights to words based on their importance.
```text
Bag of Words → word counts

TF-IDF → word importance
```

````python
from sklearn.feature_extraction.text import TfidfVectorizer

texts = [
    "I love this movie",
    "I hate this movie"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(texts)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF representation:")
print(X.toarray())
````

## Training Data

A machine learning model needs examples from which it can learn.

For sentiment classification, our training data could look like:

| Text | Label |
|---|---|
| I loved this movie | Positive |
| This movie was amazing | Positive |
| I hated this movie | Negative |

The text is the input.

The label is the expected output.

## X and y

In machine learning, we usually represent our data as:

### After feature extraction:
```text

 ↓

 ↓

````python
#Create a small dataset
texts = [
    "I loved this movie",
    "This movie was amazing",
    "I really enjoyed the film",
    "The movie was fantastic",
    "I hated this movie",
    "This movie was terrible",
    "I disliked the film",
    "The movie was boring"
]

labels = [
    "Positive",
    "Positive",
    "Positive",
    "Positive",
    "Negative",
    "Negative",
    "Negative",
    "Negative"
]

print("Number of texts:", len(texts))
print("Number of labels:", len(labels))
````

## Training and Testing Data

We should not train and test the model using exactly the same data.

Instead, we divide our dataset into:

Training data
→ Used to learn patterns

Testing data
→ Used to evaluate the model on unseen examples

A common split is:

80% → Training

20% → Testing

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    texts,
    labels,
    test_size=0.25,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
````

## Important

The model should learn from the training data.

The test data should represent new/unseen examples.

This helps us determine whether the model has actually learned useful patterns rather than simply memorizing the training examples.

````python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X_train_features = vectorizer.fit_transform(X_train)
X_test_features = vectorizer.transform(X_test)

print("Training feature shape:", X_train_features.shape)
print("Testing feature shape:", X_test_features.shape)
````

### Very important

#### Notice:
```text
fit_transform(X_train)
```
#### but:
```text
transform(X_test)
```
We fit the vectorizer only on training data.

This avoids allowing information from the test set to influence the feature-building process.

````python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X_train_features, y_train)

print("Model training completed!")
````

The model has now learned patterns from our training examples.

````python
predictions = model.predict(X_test_features)

print("Actual labels:")
print(y_test)

print("\nPredicted labels:")
print(predictions)
````

The model compares the numerical patterns it learned with the new text features.

## Predict Completely New Text

Now we can give the model a sentence that it has never seen before.

Example:

"This movie was wonderful"

The model should try to classify it as:

Positive

````python
new_text = ["This movie was wonderful"]

new_features = vectorizer.transform(new_text)

prediction = model.predict(new_features)

print("Text:", new_text[0])
print("Prediction:", prediction[0])
````

Depending on the very small training dataset, the prediction may not always be correct.

That is normal.

Real NLP models need much larger and more representative datasets.

## Evaluating a Text Classification Model

After training the model, we need to determine how well it performs.

Common metrics include:

- Accuracy
- Precision
- Recall
- F1-score
- Confusion Matrix

We will study these metrics in more detail later.

For now:

Accuracy:

How many predictions were correct?

Example:

100 predictions

90 correct

Accuracy = 90%

````python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", accuracy)
````

Because our dataset is extremely small, do not treat this accuracy as a meaningful real-world model score.

The purpose is to understand the classification workflow.

````python
from sklearn.metrics import classification_report

print(classification_report(y_test, predictions, zero_division=0))
````

## Confusion Matrix

A confusion matrix shows how predictions compare with actual classes.

For binary sentiment classification:

                Predicted
              Positive Negative

Actual
Positive        TP       FN

Negative        FP       TN


TP = True Positive
TN = True Negative
FP = False Positive
FN = False Negative

## Important Limitation of Traditional Text Classification

Traditional approaches such as Bag of Words and TF-IDF are useful, but they have limitations.

For example:

"I like this movie"

and

"I do not like this movie"

contain many of the same words.

Simple word-based representations may struggle to fully understand the meaning and relationship between words.

They also have difficulty understanding deeper context.

This is one reason NLP evolved toward:

Word embeddings
        ↓
Neural Networks
        ↓
RNNs / LSTMs
        ↓
Transformers
        ↓
Large Language Models
