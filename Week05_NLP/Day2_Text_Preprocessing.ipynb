## 1. What is Text Preprocessing?

Text preprocessing is the process of converting raw text into a cleaner and more consistent form before giving it to an NLP model.

Raw text can contain:

- Uppercase and lowercase letters
- Punctuation
- Numbers
- Special characters
- Extra spaces
- URLs
- Unnecessary symbols
- Stop words

A machine learning model generally works better when the text is prepared consistently.

### Basic Pipeline
```text
Raw Text
↓
Lowercase
↓
Remove punctuation
↓
Remove unnecessary characters
↓
Remove stop words when appropriate
↓
Clean Text

## 2. Why Do We Need Text Preprocessing?

Consider these sentences:

```text
"I LOVE this product!!!"

"I love this product."

"  I love this product   "
```
A human can easily understand that they are very similar.

However, a computer may initially treat differences such as uppercase letters, punctuation, and extra spaces as different text patterns.

Preprocessing helps make text more consistent.

### Example

Before preprocessing:
```text
"I LOVE this product!!!"
```
After basic preprocessing:
```text
"i love this product"
```
The meaning is mostly preserved while unnecessary differences are removed

## 3. Text Cleaning

Text cleaning means removing unwanted or unnecessary parts of text.

Common cleaning operations include:

- Removing extra spaces
- Removing punctuation when appropriate
- Removing unnecessary special characters
- Removing unwanted symbols
- Removing URLs when they are not useful
- Removing HTML tags when present

### Example

Raw:

```text
"Hello!!!   Welcome to my website :)"
```
After cleaning:
```text
"Hello Welcome to my website"
```
The exact cleaning steps depend on the NLP problem.

### Important

We should NOT blindly remove everything.

Some characters can carry meaning depending on the task.

For example:
```text
"I don't like this."
```
Removing the word "not" or changing the sentence incorrectly could affect sentiment analysis.

Therefore, preprocessing should be designed according to the problem.

## 5. Removing Punctuation

Punctuation marks include:

```text
. , ! ? : ; " ' ( ) [ ] { }
```
Sometimes punctuation is not necessary for the NLP task.

Example:
```text
"I love this product!!!"
```
#### can become:
```text
"I love this product"
```
### Why remove punctuation?

It can reduce unnecessary variations in the text.

For example:
```text
"hello"
"hello!"
"hello!!!"
```
can be normalized to:
```text
"hello"
```
However, punctuation can sometimes contain useful information.

Therefore, punctuation should only be removed when it makes sense for the task.

````python
# Remove Punctuation with Python

import string

text = "I love this product!!!"

clean_text = text.translate(
    str.maketrans("", "", string.punctuation)
)

print("Original:", text)
print("Cleaned:", clean_text)
````

## 9. Stop Words

Stop words are common words that may provide limited useful information for some NLP tasks.

Examples include:

```text
the
is
a
an
and
of
to
in
on
```
### For example:
```text
"The product is very good"
```
### Possible stop words:
```text
the
is
```
### Remaining words:
```text
product
very
good
```
### Why remove stop words?

### Removing common words can:
```text
Reduce the amount of text
Reduce the number of tokens/features
Make some traditional NLP tasks more efficient
```
### Important

Stop-word removal is NOT always necessary.

For tasks such as sentiment analysis, words such as:
```text
not
never
```
can be extremely important.

### For example:
```text
"This product is not good."
```
If we remove "not":
```text
"This product is good."
```
The meaning changes completely.

Therefore, stop-word removal should be used carefully.

````python
# Stop Words Example with NLTK

import nltk

nltk.download("stopwords")

from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

text="This is a very good product"

words=text.lower().split()

filtered_words=[
    word for word in words
    if word not in stop_words
]

print("original:", words)
print("After removing stop words:", filtered_words)                           
````

## 10. Complete Basic Preprocessing Pipeline

Now let's combine the concepts.

Example raw text:

```text
"  I LOVE this PRODUCT!!! 😊  "
```
We can perform:

### Step 1

Lowercase:
```text
"  i love this product!!! 😊  "
```
### Step 2

Remove punctuation:
```text
"  i love this product 😊  "
```
### Step 3

Remove unnecessary characters if appropriate:
```text
"  i love this product  "
```
### Step 4

Remove extra spaces:
```text
"i love this product"
```
Final Clean Text
```text
"i love this product"
```
### Pipeline
```text
Raw Text
↓
Lowercase
↓
Remove punctuation
↓
Remove unnecessary characters
↓
Normalize spaces
↓
Optional stop-word removal
↓
Clean Text

````python
# Build a simple cleaning function
import string

def clean_text(text):
    text=text.lower()

    text= text.translate(
    str.maketrans("","",string.punctuation)
    )
    text=" ".join(text.split())
    return text
````

````python
text = "  I LOVE this product!!!   "

cleaned = clean_text(text)

print("Original:", text)
print("Cleaned:", cleaned)
````

````python
texts = [
    "I LOVE this product!!!",
    "The product is very good.",
    "This is TERRIBLE!!!",
    "Amazing service!!!"
]

cleaned_texts = [clean_text(text) for text in texts]

for original, cleaned in zip(texts, cleaned_texts):
    print("Original :", original)
    print("Cleaned  :", cleaned)
    print()
````

## 12. Important: Don't Over-Clean Text

Text preprocessing should not destroy useful information.

### For example:

```text
"I don't like this product."
```
### If we remove "don't":
```text
"I like this product."
```
The sentiment changes from negative to positive.

Therefore, preprocessing should depend on the NLP task.

#### For our Week 5 project — Sentiment Analysis — we need to be especially careful with words such as:
```text
not
no
never
don't
isn't
can't
```
These can change the meaning of a sentence.

### Key Rule

Clean unnecessary information, but preserve information that is important for the task.
