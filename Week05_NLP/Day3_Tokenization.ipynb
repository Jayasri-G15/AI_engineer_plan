## 1. What is Tokenization?

Tokenization is the process of breaking text into smaller units called **tokens**.

A token can be:

- A word
- A sentence
- Or another meaningful piece of text

### Example

Input:

"I love machine learning"

After word tokenization:

```text
["I", "love", "machine", "learning"]
```
Each individual word becomes a token.

### Simple Definition

Tokenization is the process of splitting text into smaller units that an NLP system can process

## 2. Why Do We Need Tokenization?

A computer needs to process text in manageable pieces.

Consider:

"I love machine learning."

Instead of treating the entire sentence as one large piece of text, we can divide it into:

```text
"I"
"love"
"machine"
"learning"
```
Now each token can be processed separately.

Tokenization is often one of the early steps in an NLP pipeline.

### Basic Flow
```text
Raw Text
↓
Tokenization
↓
Individual Tokens
↓
Further NLP Processing

## 3. Word Tokenization

Word tokenization divides a sentence or document into individual words/tokens.

### Example

Sentence:

> "I love machine learning."

Result:

```text
["I", "love", "machine", "learning"]

````python
text = "I love machine learning"

words = text.split()

print(words)
````

Here we used Python's `split()` method to create a very simple word-level tokenization.

This is useful for understanding the basic idea.

However, real NLP tokenization can be more sophisticated, especially when dealing with punctuation, contractions, emojis, and other language-specific cases.

## 4. Tokenization and Punctuation

Consider:

> "I love NLP!"

A simple split gives:

```text
["I", "love", "NLP!"]
```
Notice that the ! remains attached to NLP.

More advanced tokenizers can treat punctuation separately.

#### For example:
```text
"I love NLP!"
```
#### could be represented as:
```text
["I", "love", "NLP", "!"]
```
The exact behavior depends on the tokenizer being used.

## 5. Sentence Tokenization

Sentence tokenization divides a paragraph or document into individual sentences.

Example:

```text
"I love machine learning.
NLP is interesting.
I am learning Python."
```
After sentence tokenization:
```text
[
    "I love machine learning.",
    "NLP is interesting.",
    "I am learning Python."
]
```
### Why is this useful?

It allows us to analyze text one sentence at a time.

For example, a document may contain hundreds of sentences.

We can split the document into sentences and then process each sentence separately.

````python
text = "I love machine learning. NLP is interesting. I am learning Python."

sentences = text.split(".")

for sentence in sentences:
    if sentence.strip():
        print(sentence.strip())
````

Real NLP applications usually use dedicated tokenizers because punctuation and sentence boundaries can be more complicated than a simple period split.

## 6. Tokenization Using NLTK

NLTK is a Python library commonly used for learning and working with NLP concepts.

It provides tokenization functions for words and sentences.

We can use:

```text
word_tokenize()
sent_tokenize()

````python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
````

````python
from nltk.tokenize import word_tokenize

text = "I love machine learning!"

tokens = word_tokenize(text)

print(tokens)
````

````python
from nltk.tokenize import sent_tokenize

text = "I love machine learning. NLP is interesting. I am learning Python."

sentences = sent_tokenize(text)

print(sentences)
````

## 7. Word Tokenization vs Sentence Tokenization

### Word Tokenization

Breaks text into individual words/tokens.

Example:

"I love NLP."

↓

["I", "love", "NLP", "."]

### Sentence Tokenization

Breaks a larger text into sentences.

Example:

"I love NLP. It is interesting."

↓

[
    "I love NLP.",
    "It is interesting."
]

### Easy Way to Remember

Word tokenization → **Words**

Sentence tokenization → **Sentences**

## 8. What is Stemming?

Stemming is the process of reducing words to a simpler base-like form by removing prefixes or suffixes according to a set of rules.

The result is called a **stem**.

The stem does not always have to be a valid English word.

### Example

Words such as:

```text
playing
played
plays
```
may be reduced to something similar to:
```text
play
```
### Important

Stemming is generally faster and simpler than lemmatization, but its results can sometimes be crude or not actual dictionary words.

````python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying"
]

for word in words:
    print(word, "→", stemmer.stem(word))
````

### Important Observation

Notice:
```text
studies → studi
```
"studi" is not a normal English word.

This is one limitation of stemming.

It uses rules to reduce words rather than fully understanding their linguistic meaning.

## 9. Why Can Stemming Be Crude?

A stemmer usually applies rules rather than understanding the actual meaning of the word.

For example:

```text
studies
```
may become:
```text
studi
```
Even though "studi" is not a valid English word.

Therefore:

Stemming is fast and simple, but the resulting stem may not be a real dictionary word

## 10. What is Lemmatization?

Lemmatization reduces a word to its proper dictionary/base form, called a **lemma**.

Unlike basic stemming, lemmatization attempts to consider the word's vocabulary and linguistic meaning.

### Example

```text
running → run
better → good
children → child
```
The goal is to obtain a meaningful base word.

### Simple Definition

Lemmatization reduces words to their meaningful dictionary base form.

## 11. Stemming vs Lemmatization

| Stemming | Lemmatization |
|---|---|
| Uses simple rules | Uses linguistic information |
| Usually faster | Usually more computationally expensive |
| May produce invalid words | Usually produces valid dictionary words |
| Less accurate linguistically | More linguistically meaningful |
| Good when speed is important | Good when meaning matters |

### Example

Word:

"studies"

Stemming might produce:

"studi"

Lemmatization aims for:

"study"

### Easy Memory Trick

Stemming:

> Cut the word down.

Lemmatization:

> Find the meaningful base word.

## 12. Lemmatization Using NLTK

NLTK provides a `WordNetLemmatizer`.

We first need the WordNet resource.


````python
import nltk

nltk.download("wordnet")
nltk.download("omw-1.4")
````

````python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = [
    "playing",
    "played",
    "studies",
    "cars"
]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
````

### Important

The result of lemmatization can depend on the part of speech (POS) supplied.

For example, the word "playing" may be treated differently when the system knows it is a verb.

So basic lemmatization without POS information is not always enough.

## 13. Lemmatization and Part of Speech

Words can have different grammatical roles.

For example:

```text
playing
```
can be treated as a verb.

When we provide the correct part of speech, lemmatization can produce a better base form.

Example:
```text
playing + verb
       ↓
     play
```
This shows why linguistic information can improve lemmatization.

````python
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = [
    "playing",
    "played",
    "studies",
    "cars"
]

for word in words:
    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word)
    
    print(
        f"{word:10} → Stem: {stem:10} | Lemma: {lemma}"
    )
````

The exact output may vary depending on the NLP library and the information supplied to the lemmatizer.

The important concept is:
```text
Stemming → rule-based reduction

Lemmatization → meaningful dictionary-based reduction

## 14. Why Are Stemming and Lemmatization Useful?

Consider:

```text
play
plays
played
playing
```
These words are related.

Without normalization, a model may treat them as different tokens.

Stemming or lemmatization can reduce variations toward a common base.

This can:
```
Reduce vocabulary size
Reduce redundant word forms
Help some traditional NLP models
Make text representations more consistent
```
However, they are not mandatory for every modern NLP system.

The appropriate preprocessing depends on the model and task.

## 15. Example: Sentiment Analysis

Suppose we have:

```
These sentences contain different forms of the word:
```text

```
#### The word:

is important.

So preprocessing should preserve important meaning.
